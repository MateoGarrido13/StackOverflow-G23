En principio, mantendremos estas 5 categorias determindads por `AIAcc` y evaluaremos si existen diferencias generales entre ellas a traves de un Test de Kruskal-Wallis.
Hipótesis Nula ($H_0$): Las distribuciones de los anos de experiencia son idénticas para los distintos grupos determinados por el nivel de confianza en la IA.
Hipótesis Alternativa ($H_1$): Al menos uno de los 5 grupos presenta una distribución diferente a los demás.

```python
# A priori comparamos con un boxplot por grupo; ylabel es WorkExp y xlabel el nivel de confianza en la IA.
import matplotlib.pyplot as plt

orden_confianza = [
    "Highly trust",
    "Somewhat trust",
    "Neither trust nor distrust",
    "Somewhat distrust",
    "Highly distrust",
]

grafico = df_confianza.dropna(subset=["WorkExp", "AIAcc"])
datos = [
    grafico.loc[grafico["AIAcc"] == nivel, "WorkExp"].to_numpy()
    for nivel in orden_confianza
]

plt.figure(figsize=(10, 5))
plt.boxplot(datos, vert=True)
plt.xticks(range(1, len(orden_confianza) + 1), orden_confianza, rotation=20, ha="right")
plt.ylabel("WorkExp")
plt.xlabel("Nivel de confianza en la IA")
plt.tight_layout()
plt.show()
```

Visualmente no se aprecian diferencias marcadas entre las cajas. Antes del Kruskal-Wallis se chequean sus supuestos. El test compara la distribución de `WorkExp` entre los cinco niveles de `AIAcc`.

**La variable respuesta es ordinal o cuantitativa discreta.** Kruskal-Wallis trabaja con rangos, así que la variable que se compara tiene que poder ordenarse. Aquí esa variable es `WorkExp`: años enteros de experiencia profesional (QID14), cuantitativa discreta.

**Independencia de las observaciones.** La distribución del estadístico bajo H0 supone que cada fila aporta una observación que no depende de las demás. Si la misma persona apareciera dos veces, o si varias filas fueran mediciones repetidas del mismo desarrollador, los rangos dejarían de ser intercambiables y el *p*-valor no se podría leer. En esta encuesta a cada desarrollador que respondio, se lo correpsonde con una unica muestra. 

**Misma forma y misma dispersión en los cinco grupos.** H0, tal como quedó planteada, dice que las cinco distribuciones de `WorkExp` son idénticas. Para ese enunciado, Kruskal-Wallis no exige varianzas iguales, pero de respetarla el rechazo de $H_0$ implica formalmente diferencias en la localización de los grupos (medianas/rangos). Si no lo son, el test solo indica que las distribuciones difieren en su forma general o varianza (ante un rechazo de H0).
Mediante un Test de Levene comprobaremos la homocedasticidad entre los grupos.




```python


from scipy import stats

orden_confianza = [
    "Highly trust",
    "Somewhat trust",
    "Neither trust nor distrust",
    "Somewhat distrust",
    "Highly distrust",
]

muestra = df_confianza.dropna(subset=["WorkExp", "AIAcc"])
grupos = [
    muestra.loc[muestra["AIAcc"] == nivel, "WorkExp"]
    for nivel in orden_confianza
]

# Centro en la mediana: WorkExp es asimétrica y este centro no se corre con los extremos.
stat_lev, p_lev = stats.levene(*grupos, center="median")
print(f"Levene (homocedasticidad de WorkExp entre niveles de AIAcc): estadístico = {stat_lev:.4f}, p-value = {p_lev:.4g}")
for nivel, grupo in zip(orden_confianza, grupos):
    print(f"{nivel}: n = {len(grupo)}, desvío = {grupo.std(ddof=1):.2f}")

if p_lev > 0.05:
    print("Las dispersiones son compatibles. Un rechazo de H0 en Kruskal-Wallis se leera como diferencia de localización.")
else:
    print("Las dispersiones difieren. Un rechazo de H0 en Kruskal-Wallis se leera como diferencia de distribución, no solo de mediana.")

```

```python
# 2. TEST DE KRUSKAL-WALLIS (OMNIBÚS)
# -------------------------------------------------------------
h_stat, p_val_kw = stats.kruskal(*grupos)
print(f"\n[Test de Kruskal-Wallis] H-est = {h_stat:.4f}, p-value = {p_val_kw:.4f}")
```


**CONCLUSION:** Observemos que p-valor < 0.05 (nivel de significancia), por lo tanto, los grupos presentan diferencias estadisticamente significativas y concluiriamos en que los años de experiencia no es la misma en los cinco niveles de confianza en la IA. Al menos un nivel de `AIAcc` concentra la experiencia de otra manera.
Ahora, en términos reales, las magnitudes son superficiales en el dominio. Las medianas quedan entre 10 y 12 años y el rango intercuartílico, entre 14 y 15. El *p* chico aparece por el tamaño de la muestra, no porque la experiencia cambie de nivel de confianza.
En resumen, la experiencia no necesariamente se traduce en una mayor desconfianza en la IA, debido a un criterio formado por el desarrollador, sin qoe, en el caso de roles 'entry', es posible que antes de avanzar con una implementacion sugerida por la misma, haga mas de las preguntas normales para comprender el concepto que habra por dentras de una funcion declarada, la importacion de una libreria en especifico,etc.

### Hipótesis 3: Están los distintos niveles de satisfacción en los distintos aspectos del trabajo actual relacionados a la consideración de un cambio de rol o de carrera

Con esta hipótesis se evaluara si existe, en un empleado, una relación entre la idea de cambiar de rol/puesto de trabajo y los aspectos que más aprecia de su trabajo actual. A partir de este análisis, se podría identificar un patron de satisfacción de aquellos empleados que esten pensando en presentar una renuncia para implementar mejoras específicas en el ambiente laboral. 

Hipótesis planteadas:

- Hipótesis Nula ($H_0$): Los niveles de satisfacción en los aspectos del trabajo actual no tienen relación con la probabilidad de considerar un cambio de rol/carrera ($\beta_1 = \beta_2 = \dots = \beta_{14} = 0$).
- Hipótesis Alternativa ($H_1$): Al menos uno de los aspectos de satisfacción laboral influye significativamente en la probabilidad de considerar un cambio de rol/carrera (al menos un $\beta_j \neq 0$).

Decidimos utilizar una regresión logística binaria. Debido a que la variable `NewRole` posee cinco valores diferentes, los agrupamos en dos grupos. Quien cambió de carrera, aunque haya sido de forma involuntaria, queda del lado de la intención de cambio: el único 0 es no haber considerado ni transitado.

- Con intención de cambio (`y = 1`): I have strongly considered changing my career and/or the industry I work in; I have somewhat considered changing my career and/or the industry I work in; I have transitioned into a new career and/or industry voluntarily; I have transitioned into a new career and/or industry involuntarily.

- Sin intención de cambio (`y = 0`): I have neither consider or transitioned into a new career or industry.

Si bien el cambio involuntario no nace de una decisión propia, observamos que cambiar se termina considerando y efectuando: la persona ya transitó a otra carrera o industria. Por eso esa respuesta entra en `y = 1`. En términos prácticos, la precisión de “con intención de cambio” es mayor bajo esta agrupación: 54,9% en el test, contra 52,9% si el cambio involuntario se codifica como ausencia de intención.

Teniendo en cuenta estas dos categorias, el modelo intentará predecir a partir de los aspectos de satisfacción contenidos en las variables `JobSatPoints_*` el grupo de `NewRole` al que el trabajador pertenece.

Asociamos valores 0 (Ninguna intención de cambio) y 1 (Intencion de cambio) para cada grupo dentro de la columna `NewRole`.

Se consideran como validas para el analisis todas las filas que tengan las variables `NewRole` y `JobSatPoints_*` como no nulas. El ranking debe estar completo, con respuestas numericas de 1 al 15 sin repetir.

```python
import pandas as pd

# Se asume que df_encuestas contiene los datos de entrada.
columnas_ranking = sorted(
    [
        columna for columna in df_encuestas.columns
        if columna.startswith("JobSatPoints_")
        and columna.rsplit("_", 1)[-1].isdigit()
    ],
    key=lambda columna: int(columna.rsplit("_", 1)[-1]),
)

# Conservar solo los 15 rangos completos (permutación de 1 a 15) y las cinco
# respuestas válidas de NewRole; JobSatPoints_15_TEXT no forma parte del ranking.
# 0 es únicamente quien no consideró ni transitó. Considerar el cambio, o haberlo
# hecho de forma voluntaria o involuntaria, cuenta como intención de cambio (1).
# El cambio involuntario ya se efectuó: en la práctica esta agrupación sube la
# precisión de "con intención" (54,9% contra 52,9% si ese caso queda en 0).
CATEGORIAS_CAMBIO = {
    "I have strongly considered changing my career and/or the industry I work in": 1,
    "I have somewhat considered changing my career and/or the industry I work in": 1,
    "I have transitioned into a new career and/or industry voluntarily": 1,
    "I have transitioned into a new career and/or industry involuntarily": 1,
    "I have neither consider or transitioned into a new career or industry": 0,
}

columnas_requeridas = ["NewRole", *columnas_ranking]
df_modelo = df_encuestas.loc[:, columnas_requeridas].copy()
df_modelo[columnas_ranking] = df_modelo[columnas_ranking].apply(
    pd.to_numeric, errors="coerce"
)
ranking_valido = (
    df_modelo[columnas_ranking].notna().all(axis=1)
    & df_modelo[columnas_ranking].ge(1).all(axis=1)
    & df_modelo[columnas_ranking].le(15).all(axis=1)
    & df_modelo[columnas_ranking].nunique(axis=1).eq(15)
    & df_modelo[columnas_ranking].sum(axis=1).eq(120)
)
newrole_valido = df_modelo["NewRole"].isin(CATEGORIAS_CAMBIO)
df_modelo = df_modelo.loc[ranking_valido & newrole_valido].copy()
df_modelo["y"] = df_modelo["NewRole"].map(CATEGORIAS_CAMBIO).astype(int)

print(f"Observaciones válidas: {len(df_modelo):,}")
```

```python
# Nulos de AIAgents dentro de la poblacion (profesion y empleo).
# AISelect se usa para ver si hay una moda con la que imputar, y despues se descarta.

empleos = [
    "Employed",
    "Independent contractor, freelancer, or self-employed",
]
orden_org = [
    "Just me - I am a freelancer, sole proprietor, etc.",
    "Less than 20 employees",
    "20 to 99 employees",
    "100 to 499 employees",
    "500 to 999 employees",
    "1,000 to 4,999 employees",
    "5,000 to 9,999 employees",
    "10,000 or more employees",
    # no consideramos a aquellos que no conocen el tamano de su empresa sera NaN tambien
]

orden_select = [
    "Yes, I use AI tools daily",
    "Yes, I use AI tools weekly",
    "Yes, I use AI tools monthly or infrequently",
    "No, but I plan to soon",
    "No, and I don't plan to",
]
orden_agents = [
    "Yes, I use AI agents at work daily",
    "Yes, I use AI agents at work weekly",
    "Yes, I use AI agents at work monthly or infrequently",
    "No, I use AI exclusively in copilot/autocomplete mode",
    "No, but I plan to",
    "No, and I don't plan to",
]


respondio_emp_no_agents = (
    df_interes["OrgSize"].notna()
    & df_interes["OrgSize"].isin(orden_org)
    & df_interes["AISelect"].notna()
    & df_interes["AISelect"].isin(usa_ia)
    & df_interes["AIAgents"].isna()
).sum()
print(f"Hay {respondio_emp_no_agents} valores que usan AI pero que no respondieron por el uso de agentes")

```

En un proceso similar a el de la hipotesis 2 propuesta por la catedra, aqui evaluaremos una relacion entre 'AIAgents' y 'AISelect', de forma tal que, el uso/ la frecuencia de uso de AI podria traducirse en la frecuencia de uso de Agentes.
Por ello, construimos una matriz de contongencia para evaluar si existen relaciones claras entre las mismas que permitan realizar imputaciones de valores nulos, siendo que estas dos variables podrian estar aportando informacion redundante.

```python
# MAtriz de contingencia, evaluamos frecuencia de uso de IA('AISelect') con frecuencia de uso de Agentes('AIAgents') 
respondieron_agentes= (
    df_interes["AISelect"].isin(orden_select) & df_interes["AIAgents"].notna()
)

contingencia_ai = pd.crosstab(
    df_interes.loc[respondieron_agentes, "AISelect"],
    df_interes.loc[respondieron_agentes, "AIAgents"],
).reindex(index=orden_select, columns=orden_agents)


display(contingencia_ai)
```

Al igual que en el caso anterior, las unicas correspondencias claras son en la interseccion entre un uso diario de IA con un uso diario de Agentes de IA y/o el caso opuesto de negacion a su uso; el resto de valores seran descartados del dataset a construir

```python
# Aplicaremos la transformacion de dichos valores
df_emp_agentes = df_interes[["AISelect", "AIAgents", "OrgSize"]].copy()
# procedemos a aplicar las imputaciones sobre "AIAgents"
imputar_diario = (
    df_emp_agentes["AIAgents"].isna()
    & (df_emp_agentes["AISelect"] == "Yes, I use AI tools daily")
)
imputar_no_planea = (
    df_emp_agentes["AIAgents"].isna()
    & (df_emp_agentes["AISelect"] == "No, and I don't plan to")
)
df_emp_agentes.loc[imputar_diario, "AIAgents"] = "Yes, I use AI agents at work daily"
df_emp_agentes.loc[imputar_no_planea, "AIAgents"] = "No, and I don't plan to"

print(f"Imputadas con uso diario de agentes: {int(imputar_diario.sum())}")
print(f"Imputadas con no planea usar agentes: {int(imputar_no_planea.sum())}")
```

```python
# Quedan fuera los nulos y quienes respondieron que no conocen el tamano.
no_sabe = "I don’t know"
fuera_de_poblacion = (
    df_emp_agentes["AIAgents"].isna()
    | df_emp_agentes["OrgSize"].isna()
    | (df_emp_agentes["OrgSize"] == no_sabe)
)
valores_a_eliminar = int(fuera_de_poblacion.sum())
print(valores_a_eliminar)
valores_a_analizar = len(df_emp_agentes) - valores_a_eliminar
print(f"Nos quedaremos con el {valores_a_analizar / len(df_emp_agentes) * 100}% de los datos")
# eliminaremos 'AISelect' puesto que ya realizamos las imputaciones necesarias
df_emp_agentes = df_emp_agentes.loc[~fuera_de_poblacion].drop(columns="AISelect")
df_emp_agentes.head()
```

Antes del chi-cuadrado se chequean sus supuestos. El test evalúa si `OrgSize` y `AIAgents` son independientes. Las dos variables son cualitativas. `AIAgents` analiza. frecuencia y modo de uso.
**Independencia de las observaciones.** Cada fila es una persona que respondió la encuesta. Si la misma persona apareciera dos veces, las celdas dejarían de ser intercambiables y el *p*-valor no se podría leer. En esta encuesta a cada desarrollador que respondió le corresponde una única muestra.

**Una sola celda por persona.** `OrgSize` y `AIAgents` son de opción múltiple con una respuesta. Cada fila de `df_emp_agentes` cae en un nivel de cada variable.

**Frecuencias esperadas.** La aproximación del chi-cuadrado pide que las frecuencias esperadas no sean chicas. El criterio que se usa acá: todas las celdas con esperada de al menos 5. Si alguna quedara por debajo de 1, esa categoría se agruparía. No se imputan nulos para inflar una celda.

**Tamaño de la muestra.** Con decenas de miles de filas, un *p* chico puede aparecer con una asociación débil. Por eso el contraste informa también el porcentaje de cada tamaño que declara agentes. El chi-cuadrado no dice si las organizaciones chicas usan más agentes: esa lectura sale de ese porcentaje.

```python
contingencia_org = (
    pd.crosstab(df_emp_agentes["OrgSize"], df_emp_agentes["AIAgents"])
    .reindex(index=orden_org, columns=orden_agents)
    .fillna(0)
    .astype(int)
)
display(contingencia_org)
```

Vemos que no se nota una correspondencia directa entre tamano de empresa y uso de agemtes, pero si, una marcada presencia de valores negativos al uso de agentes independientemente del tamano de la empresa, por lo tanto, haremos agrupaciones semanticas reduciendo a 3 categorias del uso de agentes (Yes, No, I use AI exclusively in copilot/autocomplete mode, No).


```python
# realizamos la agrupacion (Yes, "No, I use AI exclusively in copilot/autocomplete mode", No) sobre la variable AIAgents
# Yes: "Yes, I use AI agents at work daily", "Yes, I use AI agents at work weekly", "Yes, I use AI agents at work monthly or infrequently"
# "No, I use AI exclusively in copilot/autocomplete mode" se mantiene
# No: "No, but I plan to", "No, and I don't plan to"
agrupacion_agentes = {
    "Yes, I use AI agents at work daily": "Yes",
    "Yes, I use AI agents at work weekly": "Yes",
    "Yes, I use AI agents at work monthly or infrequently": "Yes",
    "No, only use IA across LLM": "No, I use AI exclusively in copilot/autocomplete mode",
    "No, but I plan to": "No",
    "No, and I don't plan to": "No",
}
df_emp_agentes["AIAgents"] = df_emp_agentes["AIAgents"].replace(agrupacion_agentes)
df_emp_agentes["AIAgents"].value_counts()
```

```python
from scipy import stats
# realizamos una prueba sobre las categorias existentes
chi2_org, p_org, dof_org, _ = stats.chi2_contingency(contingencia_org)
print(f"Chi-cuadrado = {chi2_org:.4f}, gl = {dof_org}, p-value = {p_org:.4g}")
print(f"n = {int(n_org)}")
```

```python
orden_agentes_agrupado = [
    "Yes",
    "No, I use AI exclusively in copilot/autocomplete mode",
    "No",
]
contingencia_org = (
    pd.crosstab(df_emp_agentes["OrgSize"], df_emp_agentes["AIAgents"])
    .reindex(index=orden_org, columns=orden_agentes_agrupado)
    .fillna(0)
    .astype(int)
)
display(contingencia_org)
```

```python
import numpy as np

tabla_org = contingencia_org.to_numpy()
n_org = tabla_org.sum()
esperadas_org = np.outer(tabla_org.sum(axis=1), tabla_org.sum(axis=0)) / n_org
esperadas_org_df = pd.DataFrame(
    esperadas_org,
    index=contingencia_org.index,
    columns=contingencia_org.columns,
)

min_esperada = esperadas_org.min()
celdas_bajo_5 = int((esperadas_org < 5).sum())
print(f"Frecuencia esperada minima: {min_esperada:.1f}")
print(f"Celdas con esperada menor que 5: {celdas_bajo_5} de {esperadas_org.size}")
display(esperadas_org_df.round(1))
```

Hipótesis nula ($H_0$): `OrgSize` y `AIAgents` son independientes. La proporción de cada categoría de `AIAgents` (`Yes`, `No, I use AI exclusively in copilot/autocomplete mode` y `No`) es la misma en todos los tamaños de organización.

Hipótesis alternativa ($H_1$): `OrgSize` y `AIAgents` no son independientes. El perfil de `AIAgents` cambia según `OrgSize`.

El test es un chi-cuadrado de independencia, con nivel de significancia 0,05.

```python
from scipy import stats
# una vez agrupadas probaremos 
chi2_org, p_org, dof_org, _ = stats.chi2_contingency(contingencia_org)
print(f"Chi-cuadrado = {chi2_org:.4f}, gl = {dof_org}, p-value = {p_org:.4g}")
print(f"n = {int(n_org)}")
```

**Conclusión.** Antes de agrupar, con las seis respuestas de `AIAgents`, el chi-cuadrado es 138,70, con 35 grados de libertad y p-valor $2{,}8 \times 10^{-14}$. También queda por debajo de 0,05: se rechaza $H_0$. La frecuencia esperada mínima es 97,2 y ninguna de las 48 celdas queda por debajo de 5.

Después de reunir las frecuencias de uso en `Yes` y las dos respuestas negativas en `No`, el chi-cuadrado es 42,66, con 14 grados de libertad y p-valor $9{,}7 \times 10^{-5}$. Se vuelve a rechazar $H_0$. La esperada mínima es 174,5 y ninguna de las 24 celdas queda por debajo de 5.

El *p* chico aparece por las 25.654 filas. Quienes trabajan solos quedan en `Yes` en el 38,6% de los casos. En las organizaciones de menos de 20 personas ese porcentaje es 35,1%. En las de 10.000 o más, 31,9%. Entre el freelancer y la organización grande hay unos 7 puntos. El gradiente va en la dirección esperada y es chico.

Juntar diario, semanal y mensual en un solo `Yes` borra las diferencias que hay entre esas tres respuestas. El uso diario concentra cerca del 55% de la variabilidad que veía el primer test, aunque esa categoría es cerca del 17% de las respuestas. El semanal aporta menos del 4% de esa variabilidad y el mensual cerca del 14%; este último, además, no sigue la misma dirección. Al quedar los tres en una sola celda, el chi-cuadrado solo compara quién usa agentes contra el resto y deja de ver esa variación interna. Por eso el estadístico baja y el *p* se aleja de cero, aunque el test sigue rechazando la independencia.

Agrupar respuestas borra el desvío interno de las categorías más extremas (uso diario) y reduce los grados de libertad.
