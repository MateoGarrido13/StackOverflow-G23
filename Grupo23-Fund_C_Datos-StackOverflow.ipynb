```python
# Agrupamos HaveWorked: una sola columna con todos los ítems, separados por ';'
# - Celdas NaN se ignoran
# - Cada celda puede traer varios ítems separados por ';' (se parten y se recortan espacios)
# DevEnvsChoice no está en df_interes. Entran todas las *HaveWorkedWith del subconjunto.
HaveWorked_cols = [
    "LanguageHaveWorkedWith",
    "DatabaseHaveWorkedWith",
    "PlatformHaveWorkedWith",
    "WebframeHaveWorkedWith",
    "DevEnvsHaveWorkedWith",
    "SOTagsHaveWorkedWith",
    "OfficeStackAsyncHaveWorkedWith",
    "CommPlatformHaveWorkedWith",
    "AIModelsHaveWorkedWith",
]


def _tokens_desde_celda(valor):
    """Lista de strings no vacíos; NaN u otro faltante -> lista vacía."""
    if pd.isna(valor):
        return []
    texto = str(valor).strip()
    if not texto:
        return []
    return [parte.strip() for parte in texto.split(";") if parte.strip()]


def _have_worked_agregado(fila):
    ordenados = []
    vistos = set()
    for col in HaveWorked_cols:
        for tok in _tokens_desde_celda(fila[col]):
            if tok not in vistos:
                vistos.add(tok)
                ordenados.append(tok)
    return ";".join(ordenados) if ordenados else pd.NA


df_interes["HaveWorkedWith"] = df_interes[HaveWorked_cols].apply(
    _have_worked_agregado, axis=1
)
df_interes[["HaveWorkedWith"]]
```

**Transformaciones previas.** `ConvertedCompYearly` ya quedó en [0, 900.000]; lo que cae afuera es NaN. Esta hipótesis no repite esa cota. Al 1/9/2026, el SMVM es de $383.000. El triple de ese valor es $1.149.000.

Se emplearía un test t de Student si se verificara lo siguiente:

- La variable posee una distribución normal, lo que se evalúa con la prueba de Shapiro-Wilk.
- La variable es cuantitativa continua y está en escala. Esta condición se cumple.
- Independencia: las observaciones de cada desarrollador son independientes.

- H0: μ(salario) ≥ 1.149.000
- H1: μ(salario) < 1.149.000

El contraste es univariado: el salario de esa población frente a un umbral que no está en el CSV y que se obtuvo de datos del Ministerio de Capital Humano de la Nación Argentina (https://www.argentina.gob.ar/trabajo/consejodelsalario).

El salario mensual comparable es `CompTotal / 12` cuando la moneda es el peso argentino (ARS); en caso contrario, se aplica una transformación.
https://www.argentina.gob.ar/trabajo/consejodelsalario

En las 90 filas en pesos que tienen ambos montos, `CompTotal / ConvertedCompYearly` da una mediana de 1.172,26 ARS por USD: es la conversión que Stack Overflow aplicó el 25 de junio de 2025. Para quien declaró dólares, el mensual en pesos es `ConvertedCompYearly × 1.172,26 / 12`. `ConvertedCompYearly` permite auditar esa cuenta.

Población del contraste: `Country = Argentina` y `CompTotal` por encima del piso. Si el monto está en pesos, debe superar 10.000. Si está en dólares, debe superar 10. En principio, mantendremos estas 5 categorias determindads por `AIAcc` y evaluaremos si existen diferencias generales entre ellas a traves de un Test de Kruskal-Wallis.
Hipótesis Nula ($H_0$): Las distribuciones de los anos de experiencia son idénticas para los distintos grupos determinados por el nivel de confianza en la IA.
Hipótesis Alternativa ($H_1$): Al menos uno de los 5 grupos presenta una distribución diferente a los demás.

```python
# A priori comparamos con un boxplot por grupo; ylabel es WorkExp y xlabel el nivel de confianza en la IA.
import matplotlib.pyplot as plt

orden_confianza = [
    "Highly trust",
    "Somewhat trust",
    "Neither trust nor distrust",
    "Somewhat distrust",
    "Highly distrust",
]

grafico = df_confianza.dropna(subset=["WorkExp", "AIAcc"])
datos = [
    grafico.loc[grafico["AIAcc"] == nivel, "WorkExp"].to_numpy()
    for nivel in orden_confianza
]

colores_confianza = {
    "Highly trust": "#2166AC",
    "Somewhat trust": "#67A9CF",
    "Neither trust nor distrust": "#D9D9D9",
    "Somewhat distrust": "#EF8A62",
    "Highly distrust": "#B2182B",
}

plt.figure(figsize=(10, 5))
cajas = plt.boxplot(datos, orientation="vertical", patch_artist=True)
for caja, nivel in zip(cajas["boxes"], orden_confianza):
    caja.set_facecolor(colores_confianza[nivel])
for mediana in cajas["medians"]:
    mediana.set_color("#1A202C")
plt.xticks(range(1, len(orden_confianza) + 1), orden_confianza, rotation=20, ha="right")
plt.ylabel("WorkExp")
plt.xlabel("Nivel de confianza en la IA")
plt.tight_layout()
plt.show()
```

Visualmente no se aprecian diferencias marcadas entre las cajas. Antes del Kruskal-Wallis se chequean sus supuestos. El test compara la distribución de `WorkExp` entre los cinco niveles de `AIAcc`.

**La variable respuesta es ordinal o cuantitativa discreta.** Kruskal-Wallis trabaja con rangos, así que la variable que se compara tiene que poder ordenarse. Aquí esa variable es `WorkExp`: años enteros de experiencia profesional (QID14), cuantitativa discreta.

**Independencia de las observaciones.** La distribución del estadístico bajo H0 supone que cada fila aporta una observación que no depende de las demás. Si la misma persona apareciera dos veces, o si varias filas fueran mediciones repetidas del mismo desarrollador, los rangos dejarían de ser intercambiables y el *p*-valor no se podría leer. En esta encuesta a cada desarrollador que respondio, se lo correpsonde con una unica muestra. 

**Misma forma y misma dispersión en los cinco grupos.** H0, tal como quedó planteada, dice que las cinco distribuciones de `WorkExp` son idénticas. Para ese enunciado, Kruskal-Wallis no exige varianzas iguales, pero de respetarla el rechazo de $H_0$ implica formalmente diferencias en la localización de los grupos (medianas/rangos). Si no lo son, el test solo indica que las distribuciones difieren en su forma general o varianza (ante un rechazo de H0).
Mediante un Test de Levene comprobaremos la homocedasticidad entre los grupos.




```python


from scipy import stats

orden_confianza = [
    "Highly trust",
    "Somewhat trust",
    "Neither trust nor distrust",
    "Somewhat distrust",
    "Highly distrust",
]

muestra = df_confianza.dropna(subset=["WorkExp", "AIAcc"])
grupos = [
    muestra.loc[muestra["AIAcc"] == nivel, "WorkExp"]
    for nivel in orden_confianza
]

# Centro en la mediana: WorkExp es asimétrica y este centro no se corre con los extremos.
stat_lev, p_lev = stats.levene(*grupos, center="median")
print(f"Levene (homocedasticidad de WorkExp entre niveles de AIAcc): estadístico = {stat_lev:.4f}, p-value = {p_lev:.4g}")
for nivel, grupo in zip(orden_confianza, grupos):
    print(f"{nivel}: n = {len(grupo)}, desvío = {grupo.std(ddof=1):.2f}")

if p_lev > 0.05:
    print("Las dispersiones son compatibles. Un rechazo de H0 en Kruskal-Wallis se leera como diferencia de localización.")
else:
    print("Las dispersiones difieren. Un rechazo de H0 en Kruskal-Wallis se leera como diferencia de distribución, no solo de mediana.")

```

```python
# 2. TEST DE KRUSKAL-WALLIS (OMNIBÚS)
# -------------------------------------------------------------
h_stat, p_val_kw = stats.kruskal(*grupos)
print(f"\n[Test de Kruskal-Wallis] H-est = {h_stat:.4f}, p-value = {p_val_kw:.4f}")
```


**CONCLUSION:** Observemos que p-valor < 0.05 (nivel de significancia), por lo tanto, los grupos presentan diferencias estadisticamente significativas y concluiriamos en que los años de experiencia no es la misma en los cinco niveles de confianza en la IA. Al menos un nivel de `AIAcc` concentra la experiencia de otra manera.
Ahora, en términos reales, las magnitudes son superficiales en el dominio. Levene rechaza la igualdad de dispersiones, pero el desvío de `WorkExp` va de 10,30 años (Somewhat trust) a 10,86 años (Highly distrust): menos de un año de diferencia. Las medianas quedan entre 10 y 12 años y el rango intercuartílico, entre 14 y 15. El *p* chico aparece por el tamaño de la muestra, no porque la experiencia cambie de nivel de confianza.
En resumen, la experiencia no necesariamente se traduce en una mayor desconfianza en la IA, debido a un criterio formado por el desarrollador, sin qoe, en el caso de roles 'entry', es posible que antes de avanzar con una implementacion sugerida por la misma, haga mas de las preguntas normales para comprender el concepto que habra por dentras de una funcion declarada, la importacion de una libreria en especifico,etc.

### Hipótesis 3: Están los distintos niveles de satisfacción en los distintos aspectos del trabajo actual relacionados a la consideración de un cambio de rol o de carrera

Con esta hipótesis se evaluara si existe, en un empleado, una relación entre la idea de cambiar de rol/puesto de trabajo y los aspectos que más aprecia de su trabajo actual. A partir de este análisis, se podría identificar un patron de satisfacción de aquellos empleados que esten pensando en presentar una renuncia para implementar mejoras específicas en el ambiente laboral. 

Hipótesis planteadas:

- Hipótesis Nula ($H_0$): Los niveles de satisfacción en los aspectos del trabajo actual no tienen relación con la probabilidad de considerar un cambio de rol/carrera ($\beta_1 = \beta_2 = \dots = \beta_{14} = 0$).
- Hipótesis Alternativa ($H_1$): Al menos uno de los aspectos de satisfacción laboral influye significativamente en la probabilidad de considerar un cambio de rol/carrera (al menos un $\beta_j \neq 0$).

Decidimos utilizar una regresión logística binaria. Debido a que la variable `NewRole` posee cinco valores diferentes, los agrupamos en dos grupos:

- Sin intención de cambio: I have transitioned into a new career and/or industry involuntarily // I have neither consider or transitioned into a new career or industry

Teniendo en cuenta estas dos categorias, el modelo intentará predecir a partir de los aspectos de satisfacción contenidos en las variables `JobSatPoints_*` el grupo de `NewRole` al que el trabajador pertenece.

Se consideran como validas para el analisis todas las filas que tengan las variables `NewRole` y `JobSatPoints_*` como no nulas. El ranking debe estar completo, con respuestas numericas de 1 al 15 sin repetir.

```python
import pandas as pd

columnas_ranking = sorted(
    [
        columna for columna in df_interes.columns
        if columna.startswith("JobSatPoints_")
        and columna.rsplit("_", 1)[-1].isdigit()
    ],
    key=lambda columna: int(columna.rsplit("_", 1)[-1]),
)

CATEGORIAS_CAMBIO = {
    "I have strongly considered changing my career and/or the industry I work in": 1,
    "I have somewhat considered changing my career and/or the industry I work in": 1,
    "I have transitioned into a new career and/or industry voluntarily": 1,
    "I have neither consider or transitioned into a new career or industry": 0,
    "I have transitioned into a new career and/or industry involuntarily": 0,
}

columnas_requeridas = ["NewRole", *columnas_ranking]
df_modelo = df_interes.loc[:, columnas_requeridas].copy()
df_modelo[columnas_ranking] = df_modelo[columnas_ranking].apply(
    pd.to_numeric, errors="coerce"
)
ranking_valido = (
    df_modelo[columnas_ranking].notna().all(axis=1)
    & df_modelo[columnas_ranking].ge(1).all(axis=1)
    & df_modelo[columnas_ranking].le(15).all(axis=1)
    & df_modelo[columnas_ranking].nunique(axis=1).eq(15)
    & df_modelo[columnas_ranking].sum(axis=1).eq(120)
)
newrole_valido = df_modelo["NewRole"].isin(CATEGORIAS_CAMBIO)
df_modelo = df_modelo.loc[ranking_valido & newrole_valido].copy()
df_modelo["y"] = df_modelo["NewRole"].map(CATEGORIAS_CAMBIO).astype(int)

print(f"Observaciones válidas: {len(df_modelo):,}")
```

A priori, el gráfico de barras revela una marcada asimetría en la distribución de las respuestas; es evidente la prevalencia del trabajo puramente remoto.
La alternativa 'Your choice' (very flexible, you can come in when you want or just as needed) registra la menor cantidad de respuestas, ubicándose cerca de los 4.300 registros. 
Esta disparidad visual sugiere que la elección de modalidad de trabajo no se distribuye de manera uniforme entre las 5 alternativas disponibles. Para comprobar cuantitativamente si estas diferencias observadas son estadísticamente significativas y no producto del azar muestral, se procederá a realizar una prueba formal de inferencia.

Para determinar si `RemoteWork` se ajusta a una distribución uniforme (equiprobable), utilizaremos la Prueba de Chi-Cuadrado ($\chi^2$) de Bondad de Ajuste;este test comparará las frecuencias observadas en cada una de las 5 barras contra la frecuencia esperada bajo el supuesto de que el 20% de las respuestas corresponde a cada categoría ($E_i = N / 5$).

Hipótesis Nula ($H_0$): La distribución de las modalidades de trabajo en la población es equitativa (uniforme). Todas las categorías tienen la misma proporción de respuestas ($20\%$ cada una).$$H_0: p_{\text{Remote}} = p_{\text{Hybrid-InPerson}} = p_{\text{Hybrid-Flex}} = p_{\text{In-person}} = p_{\text{YourChoice}} = 0{,}20$$Hipótesis Alternativa ($H_1$): La distribución de las modalidades de trabajo no es uniforme. Al menos una de las categorías presenta una proporción poblacional diferente al $20\%$.$$H_1: \exists \, i \in \{\text{Remote, Hybrid-InPerson, Hybrid-Flex, In-person, YourChoice}\} \quad \text{tal que} \quad p_i \neq 0{,}20$$

```python
import scipy.stats as stats
import pandas as pd

N = frecuencias.sum()
k = len(frecuencias)

frecuencia_esperada_por_categoria = N / k
frecuencias_esperadas = [frecuencia_esperada_por_categoria] * k

chi2_stat, p_val = stats.chisquare(f_obs=frecuencias, f_exp=frecuencias_esperadas)

print(f"Estadístico Chi-Cuadrado (Chi2): {chi2_stat:.4f}")
print(f"p-valor: {p_val:.4e}")

if p_val < 0.05:
    print("-> Resultado: Se rechaza H0 con un nivel de significancia del 5%. Existe una tendencia no uniforme clara en la modalidad de trabajo.")
else:
    print("-> Resultado: No se rechaza H0. Las modalidades se distribuyen uniformemente.")
```

Por lo tanto, comprobamos que la modalidad remota se destaca por encima de las demás y podremos concluir que se le adjudica a ser beneficioso tanto para el desarrollador que se le otorga una mayor flexibilidad, como al empleador de no tener que solventar una gran oficina.

El test es un chi-cuadrado de independencia, con nivel de significancia 0,05.

**Conclusión.** Antes de agrupar, con las seis respuestas de `AIAgents`, el chi-cuadrado es 138,70, con 35 grados de libertad y p-valor $2{,}8 \times 10^{-14}$. También queda por debajo de 0,05: se rechaza $H_0$. La frecuencia esperada mínima es 97,2 y ninguna de las 48 celdas queda por debajo de 5.

Después de reunir las frecuencias de uso en `Yes` y las dos respuestas negativas en `No`, el chi-cuadrado es 42,66, con 14 grados de libertad y p-valor $9{,}7 \times 10^{-5}$. Se vuelve a rechazar $H_0$. La esperada mínima es 174,5 y ninguna de las 24 celdas queda por debajo de 5.

El *p* chico aparece por las 25.654 filas. Quienes trabajan solos quedan en `Yes` en el 38,6% de los casos. En las organizaciones de menos de 20 personas ese porcentaje es 35,1%. En las de 10.000 o más, 31,9%. Entre el freelancer y la organización grande hay unos 7 puntos. El gradiente va en la dirección esperada y es chico.

Juntar diario, semanal y mensual en un solo `Yes` borra las diferencias que hay entre esas tres respuestas. El uso diario concentra cerca del 55% de la variabilidad que veía el primer test, aunque esa categoría es cerca del 17% de las respuestas. El semanal aporta menos del 4% de esa variabilidad y el mensual cerca del 14%; este último, además, no sigue la misma dirección. Al quedar los tres en una sola celda, el chi-cuadrado solo compara quién usa agentes contra el resto y deja de ver esa variación interna. Por eso el estadístico baja y el *p* se aleja de cero, aunque el test sigue rechazando la independencia.

Agrupar respuestas borra el desvío interno de las categorías más extremas (uso diario) y reduce los grados de libertad.

### Hipótesis 3: Perfiles de trayectoria, herramientas y satisfacción laboral

Entre desarrolladores de profesión empleados o independientes con respuestas válidas, existen perfiles diferenciables según sus años de experiencia profesional `WorkExp`, años programando `YearsCode` y cantidad de herramientas utilizadas en el trabajo `ToolCountWork` y en proyectos personales `ToolCountPersonal`. Se espera que la distribución de satisfacción laboral `JobSat` difiera entre esos perfiles.

- Hipótesis Nula ($H_0$): No existen perfiles diferenciables de desarrolladores según sus años de experiencia y cantidad de herramientas utilizadas, y la distribución de la satisfacción laboral (`JobSat`) se mantiene homogénea e independiente de la trayectoria profesional.
- Hipótesis Alternativa ($H_1$): Existen perfiles diferenciables de desarrolladores en función de sus años de experiencia y herramientas utilizadas, y la distribución de la satisfacción laboral (`JobSat`) difiere significativamente entre dichos perfiles.


| Variable | Pregunta |
|---|---|
| `WorkExp` | QID14. Years of professional work experience |
|`YearsCode`| QID13. Years coding in total (including education) |
|`ToolCountWork`| QID30. Distinct software tools used in primary work role|
|`ToolCountPersonal`| QID31. Distinct software tools used in side / personal projects|
|`JobSat`| QID26. How satisfied are you in your current professional developer role? (0–10)|

Para este analisis las cuatro variables predictoras forman perfiles mediante PCA exploratorio y K-Means. `JobSat` se reserva para describir y contrastar los perfiles después, evitando circularidad.

**Transformaciones previas.** `WorkExp` y `YearsCode` llegan con el reemplazo por NaN de la sección Transformaciones (fuera de [0, 60]). El cálculo del recorte de los años, y el de `ToolCountWork` y `ToolCountPersonal`, está en el texto inmediatamente anterior a la celda que los aplica.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

PREDICTORS = ["WorkExp", "YearsCode", "ToolCountWork", "ToolCountPersonal"]
JOB_SAT = "JobSat"
EMPLOYMENT_VALUES = ["Employed", "Independent contractor, freelancer, or self-employed"]

columns = ["MainBranch", "Employment", *PREDICTORS, JOB_SAT]
df = df_interes[columns].copy()

population_mask = (
    df["MainBranch"].eq("I am a developer by profession")
    & df["Employment"].isin(EMPLOYMENT_VALUES)
)
df_population = df.loc[population_mask, columns].copy()

print(f"Población filtrada: {len(df_population):,} ({len(df_population) / len(df):.1%} de la encuesta)")
nulos = pd.DataFrame({
    "n_nulos": df_population[[*PREDICTORS, JOB_SAT]].isna().sum(),
    "porcentaje_nulos": df_population[[*PREDICTORS, JOB_SAT]].isna().mean().mul(100),
})
display(nulos.round(2))


for column in [*PREDICTORS, JOB_SAT]:
    df_population[column] = pd.to_numeric(df_population[column], errors="coerce").astype("float64")

complete_predictors = df_population[PREDICTORS].notna().all(axis=1)
X_raw = df_population.loc[complete_predictors, PREDICTORS].copy()
print(f"Casos completos para PCA/K-Means: {len(X_raw):,}")
print(f"Casos completos con JobSat para el contraste: {int(df_population.loc[X_raw.index, JOB_SAT].notna().sum()):,}")
```

Los casos completos son los que imprime la celda anterior. `WorkExp` y `YearsCode` ya tienen el NaN de Transformaciones. Si `JobSat` es nula, la fila no entra al contraste de satisfacción; sí puede entrar al PCA.

El bloque siguiente aplica el tratamiento de `ToolCountWork` y `ToolCountPersonal` y el recorte de los años, y explica el valor que se carga.


```python
display(X_raw.describe(percentiles=[0.5, 0.9, 0.95, 0.99]).T)

fig, axes = plt.subplots(1, len(PREDICTORS), figsize=(16, 4))
for ax, column in zip(axes, PREDICTORS):
    sns.boxplot(y=X_raw[column], ax=ax, color="skyblue")
    ax.set_title(f"{column} — antes")
    ax.set_ylabel("Valor")
plt.tight_layout()
plt.show()
```

`WorkExp` y `YearsCode` llegan con la corrección de la sección **Transformaciones**: lo que cae fuera de [0, 60] ya es NaN y no entra a los casos completos. El boxplot anterior de esas dos variables es el de esa corrección.

En esta hipótesis, `ToolCountWork` y `ToolCountPersonal` se corrigen en la celda siguiente. Un conteo fuera de [0, 20] pasa a NaN y esa fila sale de la muestra. El valor que se carga en ese paso es NaN. Sobre las filas que siguen, los conteos que superan el percentil 95 se reemplazan por la mediana de la columna, calculada después de sacar esos NaN y antes del reemplazo. En las 16.736 filas que quedan, esa mediana es **6** para `ToolCountWork` (umbral 16) y **4** para `ToolCountPersonal` (umbral 13).

Esas imputaciones se hacen porque, dentro de [0, 20], la cola derecha sigue siendo larga. Después las variables se estandarizan y esa cola agranda la varianza que usa el PCA. La fila se conserva: se cambia el número extremo.

`WorkExp` y `YearsCode` se recortan al percentil 99 sobre esas mismas 16.736 filas. El cálculo es la interpolación lineal: en la muestra ordenada de n filas, la posición es (n − 1) × 0,99, y el valor de esa posición es el que se carga en todo dato estrictamente mayor. Ese valor es **42** en `WorkExp` y **46** en `YearsCode`. La tabla de la celda siguiente imprime los umbrales y las medianas.


```python
# WorkExp y YearsCode ya traen el NaN de Transformaciones (fuera de [0, 60]).
# ToolCountWork y ToolCountPersonal: fuera de [0, 20] pasa a NaN y la fila sale de la muestra.
# Sobre lo que queda, los años se recortan al percentil 99 y los conteos por encima del percentil 95 pasan a la mediana.
X_clean = X_raw.copy()

tool_columns = ["ToolCountWork", "ToolCountPersonal"]
fuera_de_cota = pd.Series(False, index=X_clean.index)
for column in tool_columns:
    fuera = X_clean[column].notna() & ~X_clean[column].between(0, 20)
    X_clean.loc[fuera, column] = pd.NA
    fuera_de_cota = fuera_de_cota | fuera
    print(f"{column}: {int(fuera.sum())} valores fuera de [0, 20] pasan a NaN")

X_clean = X_clean.loc[~fuera_de_cota].copy()
X_raw = X_raw.loc[X_clean.index].copy()
print(f"Filas que siguen: {len(X_clean):,}")

LIMIT_QUANTILE = 0.99
cap_columns = ["WorkExp", "YearsCode"]
cap_limits = X_raw[cap_columns].quantile(LIMIT_QUANTILE)
X_clean[cap_columns] = X_raw[cap_columns].clip(upper=cap_limits, axis="columns")

IMPUTE_QUANTILE = 0.95
impute_columns = ["ToolCountWork", "ToolCountPersonal"]
impute_limits = X_raw[impute_columns].quantile(IMPUTE_QUANTILE)
impute_medians = X_raw[impute_columns].median()
imputed_counts = {}
for column in impute_columns:
    outlier_mask = X_raw[column] > impute_limits[column]
    X_clean.loc[outlier_mask, column] = impute_medians[column]
    imputed_counts[column] = int(outlier_mask.sum())

outlier_report = pd.DataFrame({
    "tratamiento": ["Recorte superior p99", "Recorte superior p99", "Imputacion por mediana sobre p95", "Imputacion por mediana sobre p95"],
    "umbral": [cap_limits["WorkExp"], cap_limits["YearsCode"], impute_limits["ToolCountWork"], impute_limits["ToolCountPersonal"]],
    "mediana_imputacion": [np.nan, np.nan, impute_medians["ToolCountWork"], impute_medians["ToolCountPersonal"]],
    "casos_afectados": [
        int(X_raw["WorkExp"].gt(cap_limits["WorkExp"]).sum()),
        int(X_raw["YearsCode"].gt(cap_limits["YearsCode"]).sum()),
        imputed_counts["ToolCountWork"],
        imputed_counts["ToolCountPersonal"],
    ],
}, index=["WorkExp", "YearsCode", "ToolCountWork", "ToolCountPersonal"])
display(outlier_report)
display(X_clean.describe(percentiles=[0.5, 0.9, 0.95, 0.99]).T)

fig, axes = plt.subplots(1, len(PREDICTORS), figsize=(16, 4))
for ax, column in zip(axes, PREDICTORS):
    sns.boxplot(y=X_clean[column], ax=ax, color="lightgreen")
    ax.set_title(f"{column} — después del tratamiento")
    ax.set_ylabel("Valor")
plt.tight_layout()
plt.show()
```

Luego de la limpieza podemos observar que los numeros maximos tienen mucho mas sentido en esos percentiles.

Como última sección, realizamos un clustering K-Means exploratorio con el objetivo de estudiar el comportamiento de las metricas en estos casos donde no es óptima su aplicación.

Primero necesitamos saber que cantidad de cluster (K) utilizar. Para esto, hacemos un grafico de codo y comparamos los resultados de Silhouette score y Davies-Bouldin score. Ninguna metrica determina or si sola el K optimo.

```python
from sklearn.metrics import davies_bouldin_score, silhouette_score
from sklearn.cluster import KMeans

K_VALUES = range(2, 11)
k_results = []
for k in K_VALUES:
    model = KMeans(n_clusters=k, random_state=42, n_init=20)
    labels = model.fit_predict(X_scaled)
    silhouette_n = len(X_scaled)
    k_results.append({
        "k": k,
        "wcss": model.inertia_,
        "silhouette": silhouette_score(
            X_scaled, labels, sample_size=silhouette_n, random_state=42
        ),
        "davies_bouldin": davies_bouldin_score(X_scaled, labels),
    })
validation = pd.DataFrame(k_results).set_index("k")
display(validation.round(4))

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(validation.index, validation["wcss"], marker="o")
ax.set(title="Método del codo", xlabel="Número de clusters (K)", ylabel="WCSS / inercia")
ax.set_xticks(list(K_VALUES))
ax.grid(alpha=0.3)
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
validation["silhouette"].plot(kind="bar", ax=axes[0], color="teal")
axes[0].set(title="Silhouette (mayor es mejor)", xlabel="K", ylabel="Silhouette")
validation["davies_bouldin"].plot(kind="bar", ax=axes[1], color="darkorange")
axes[1].set(title="Davies–Bouldin (menor es mejor)", xlabel="K", ylabel="Índice")
for ax in axes:
    ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

En relacion a la distribucion de la muestra, estos graficos no son faciles de interpretar a simple vista dada la concentracion monotona de las muestras.
La mayor pendiente en el grafico de codo va de dos a tres, aqui buscamos el punto o puntos en los que se reduzca la caida. Sin embargo, a medida que crece el numero de clusters, la tendencia de la pendiente negativa es constante, con leves reducciones. Sacar una conclusion apartir del grafico del codo es dificil. Gracias a la puntuacion de Silhouette (la mayor) y Davies-Bouldin (la menor), podemos concluir que un K = 3 seria el numero optimo de clusters para este caso irregular.

```python
SELECTED_K = 3
if SELECTED_K not in validation.index:
    raise ValueError(f"SELECTED_K debe estar entre {min(K_VALUES)} y {max(K_VALUES)}.")

kmeans = KMeans(n_clusters=SELECTED_K, random_state=42, n_init=20)
cluster_labels = pd.Series(kmeans.fit_predict(X_scaled), index=X_scaled.index, name="Cluster")
cluster_pca = pca_df.join(cluster_labels)

fig, ax = plt.subplots(figsize=(9, 6))
sns.scatterplot(data=cluster_pca, x="PC1", y="PC2", hue="Cluster", palette="tab10", alpha=0.6, s=18, ax=ax)
ax.set_title(f"Clusters K-Means proyectados en PC1–PC2 (K={SELECTED_K})")
ax.grid(alpha=0.2)
plt.show()

cluster_sizes = cluster_labels.value_counts().sort_index().rename("N").to_frame()
cluster_sizes["Proporcion"] = (cluster_sizes["N"] / len(cluster_labels)).round(3)
display(cluster_sizes)

profile_data = X_raw.join(cluster_labels)
cluster_profiles = profile_data.groupby("Cluster")[PREDICTORS].agg(["mean", "median", "std"])
display(cluster_profiles.round(2))
```

En el scatter-plot se pintaron de color las muestras segun el cluster al que pertenecen. Los grupos estan se reparten en tres zonas del grafico, con la segunta tabla podemos identificar el criterio de agrupación de los perfiles. En el grupo 0 se encuentran aquellos empleados con mayor cantidad de años de experiencia, en el 1 aquellos con mayor numero de herramientas y en el 0 los que carecen tanto de experiencia como de herramientas. El problema de esta division es la arbitrariadad de los limites. La masiva concentracion de muestras en el centro deberian pertenecer a un solo cluster dado su cercania, sin embargo vemos que estan separadas con una superposicion muy alta en los limites de cada cluster.

Para finalizar, evaluamos la diferencia entre las distribuciones de satisfaccion (`JobSat`) de cada uno de los perfiles con un test Kruskal-Wallis.
Mostramos estas distribuciones con boxplots.

```python
from scipy.stats import kruskal

contrast = pd.DataFrame({
    "Cluster": cluster_labels,
    JOB_SAT: df_population.loc[cluster_labels.index, JOB_SAT],
})
contrast = contrast.dropna(subset=[JOB_SAT])

job_sat_stats = contrast.groupby("Cluster")[JOB_SAT].agg(
    n="count",
    mean="mean",
    median="median",
    std="std",
)
quartiles = contrast.groupby("Cluster")[JOB_SAT].quantile([0.25, 0.75]).unstack()
quartiles.columns = ["q25", "q75"]
job_sat_stats["IQR"] = quartiles["q75"] - quartiles["q25"]
display(job_sat_stats.round(3))

fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(
    data=contrast,
    x="Cluster",
    y=JOB_SAT,
    hue="Cluster",
   color="skyblue",
    legend=False,
    ax=ax,
)
ax.set_title("Satisfacción laboral por cluster — boxplot")
ax.set_ylabel("JobSat (0–10)")
ax.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()

groups = [
    group[JOB_SAT].to_numpy()
    for _, group in contrast.groupby("Cluster")
    if not group.empty
]

kw_stat, kw_pvalue = kruskal(*groups)
ALPHA_JOB_SAT = 0.05
print(f"Kruskal–Wallis: H={kw_stat:.4f}, p={kw_pvalue:.6g}")
```

Este resultado con un p-valor < 0.05 indica que al menos una distribución difiere, pero no identifica qué pares difieren ni implica una diferencia sustantiva o causal. La conclusión estadística debe interpretarse junto con la calidad del clustering, el tamaño de los grupos, la varianza explicada y sus perfiles. Un p-valor pequeño no valida por sí mismo los clusters. 

A lo largo del flujo de este analisis mencionamos las diferentes complejidades que presentaron las distribuciones de las variables que derivaron en un clustering poco satisfactorio. La diferencia entre la distribucion de satisfaccion del cluster 0 (mayor cantidad de herramientas) y los dos restantes, no refleja la realidad de los datos debido a la pobre delimitación de los cluster. Dado el conjunto de evidencias expuestas, aceptamos la hipótesis nula.
