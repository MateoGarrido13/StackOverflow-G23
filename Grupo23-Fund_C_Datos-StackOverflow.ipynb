En principio, mantendremos estas 5 categorias determindads por `AIAcc` y evaluaremos si existen diferencias generales entre ellas a traves de un Test de Kruskal-Wallis.
Hipótesis Nula ($H_0$): Las distribuciones de los anos de experiencia son idénticas para los distintos grupos determinados por el nivel de confianza en la IA.
Hipótesis Alternativa ($H_1$): Al menos uno de los 5 grupos presenta una distribución diferente a los demás.

```python
# A priori comparamos con un boxplot por grupo; ylabel es WorkExp y xlabel el nivel de confianza en la IA.
import matplotlib.pyplot as plt

orden_confianza = [
    "Highly trust",
    "Somewhat trust",
    "Neither trust nor distrust",
    "Somewhat distrust",
    "Highly distrust",
]

grafico = df_confianza.dropna(subset=["WorkExp", "AIAcc"])
datos = [
    grafico.loc[grafico["AIAcc"] == nivel, "WorkExp"].to_numpy()
    for nivel in orden_confianza
]

plt.figure(figsize=(10, 5))
plt.boxplot(datos, vert=True)
plt.xticks(range(1, len(orden_confianza) + 1), orden_confianza, rotation=20, ha="right")
plt.ylabel("WorkExp")
plt.xlabel("Nivel de confianza en la IA")
plt.tight_layout()
plt.show()
```

Visualmente no se aprecian diferencias marcadas entre las cajas. Antes del Kruskal-Wallis se chequean sus supuestos. El test compara la distribución de `WorkExp` entre los cinco niveles de `AIAcc`.

**La variable respuesta es ordinal o cuantitativa discreta.** Kruskal-Wallis trabaja con rangos, así que la variable que se compara tiene que poder ordenarse. Aquí esa variable es `WorkExp`: años enteros de experiencia profesional (QID14), cuantitativa discreta.

**Independencia de las observaciones.** La distribución del estadístico bajo H0 supone que cada fila aporta una observación que no depende de las demás. Si la misma persona apareciera dos veces, o si varias filas fueran mediciones repetidas del mismo desarrollador, los rangos dejarían de ser intercambiables y el *p*-valor no se podría leer. En esta encuesta a cada desarrollador que respondio, se lo correpsonde con una unica muestra. 

**Misma forma y misma dispersión en los cinco grupos.** H0, tal como quedó planteada, dice que las cinco distribuciones de `WorkExp` son idénticas. Para ese enunciado, Kruskal-Wallis no exige varianzas iguales, pero de respetarla el rechazo de $H_0$ implica formalmente diferencias en la localización de los grupos (medianas/rangos). Si no lo son, el test solo indica que las distribuciones difieren en su forma general o varianza (ante un rechazo de H0).
Mediante un Test de Levene comprobaremos la homocedasticidad entre los grupos.




```python


from scipy import stats

orden_confianza = [
    "Highly trust",
    "Somewhat trust",
    "Neither trust nor distrust",
    "Somewhat distrust",
    "Highly distrust",
]

muestra = df_confianza.dropna(subset=["WorkExp", "AIAcc"])
grupos = [
    muestra.loc[muestra["AIAcc"] == nivel, "WorkExp"]
    for nivel in orden_confianza
]

# Centro en la mediana: WorkExp es asimétrica y este centro no se corre con los extremos.
stat_lev, p_lev = stats.levene(*grupos, center="median")
print(f"Levene (homocedasticidad de WorkExp entre niveles de AIAcc): estadístico = {stat_lev:.4f}, p-value = {p_lev:.4g}")
for nivel, grupo in zip(orden_confianza, grupos):
    print(f"{nivel}: n = {len(grupo)}, desvío = {grupo.std(ddof=1):.2f}")

if p_lev > 0.05:
    print("Las dispersiones son compatibles. Un rechazo de H0 en Kruskal-Wallis se leera como diferencia de localización.")
else:
    print("Las dispersiones difieren. Un rechazo de H0 en Kruskal-Wallis se leera como diferencia de distribución, no solo de mediana.")

```

```python
# 2. TEST DE KRUSKAL-WALLIS (OMNIBÚS)
# -------------------------------------------------------------
h_stat, p_val_kw = stats.kruskal(*grupos)
print(f"\n[Test de Kruskal-Wallis] H-est = {h_stat:.4f}, p-value = {p_val_kw:.4f}")
```


**CONCLUSION:** Observemos que p-valor < 0.05 (nivel de significancia), por lo tanto, los grupos presentan diferencias estadisticamente significativas y concluiriamos en que los años de experiencia no es la misma en los cinco niveles de confianza en la IA. Al menos un nivel de `AIAcc` concentra la experiencia de otra manera.
Ahora, en términos reales, las magnitudes son superficiales en el dominio. Levene rechaza la igualdad de dispersiones, pero el desvío de `WorkExp` va de 10,30 años (Somewhat trust) a 10,86 años (Highly distrust): menos de un año de diferencia. Las medianas quedan entre 10 y 12 años y el rango intercuartílico, entre 14 y 15. El *p* chico aparece por el tamaño de la muestra, no porque la experiencia cambie de nivel de confianza.
En resumen, la experiencia no necesariamente se traduce en una mayor desconfianza en la IA, debido a un criterio formado por el desarrollador, sin qoe, en el caso de roles 'entry', es posible que antes de avanzar con una implementacion sugerida por la misma, haga mas de las preguntas normales para comprender el concepto que habra por dentras de una funcion declarada, la importacion de una libreria en especifico,etc.

### Hipótesis 3: Están los distintos niveles de satisfacción en los distintos aspectos del trabajo actual relacionados a la consideración de un cambio de rol o de carrera

Con esta hipótesis se evaluara si existe, en un empleado, una relación entre la idea de cambiar de rol/puesto de trabajo y los aspectos que más aprecia de su trabajo actual. A partir de este análisis, se podría identificar un patron de satisfacción de aquellos empleados que esten pensando en presentar una renuncia para implementar mejoras específicas en el ambiente laboral. 

Hipótesis planteadas:

- Hipótesis Nula ($H_0$): Los niveles de satisfacción en los aspectos del trabajo actual no tienen relación con la probabilidad de considerar un cambio de rol/carrera ($\beta_1 = \beta_2 = \dots = \beta_{14} = 0$).
- Hipótesis Alternativa ($H_1$): Al menos uno de los aspectos de satisfacción laboral influye significativamente en la probabilidad de considerar un cambio de rol/carrera (al menos un $\beta_j \neq 0$).

Decidimos utilizar una regresión logística binaria. Debido a que la variable `NewRole` posee cinco valores diferentes, los agrupamos en dos grupos:

- Con intencion de cambio: I have strongly considered changing my career and/or the industry I work in // I have somewhat considered changing my career and/or the industry I work in // I have transitioned into a new career and/or industry voluntarily

- Sin intención de cambio: I have transitioned into a new career and/or industry involuntarily // I have neither consider or transitioned into a new career or industry

Teniendo en cuenta estas dos categorias, el modelo intentará predecir a partir de los aspectos de satisfacción contenidos en las variables `JobSatPoints_*` el grupo de `NewRole` al que el trabajador pertenece.

#### Variables

| Variable | Pregunta |
|---|---|
| `NewRole` | Career change or new role in the past year |
| `JobSatPoints_1` | Control over the level of quality in projects |
| `JobSatPoints_2` | Autonomy and trust to manage your own tasks |
| `JobSatPoints_16` | You like your manager |
| `JobSatPoints_3` | Collaboration and support from a team |
| `JobSatPoints_4` | Expert mentors |
| `JobSatPoints_5` | Opportunities to mentor and lead junior-level coworkers |
| `JobSatPoints_6` | Developing specialized expertise |
| `JobSatPoints_7` | Solving real-world problems |
| `JobSatPoints_8` | Job stability and career growth with a single employer |
| `JobSatPoints_9` | Innovation through solving challenging and complex problems |
| `JobSatPoints_10` | Working with new technologies and tools |
| `JobSatPoints_11` | Competitive pay and benefits |
| `JobSatPoints_13` | Recognition from peers for your work |
| `JobSatPoints_14` | Recognition from leadership for your work |
| `JobSatPoints_15` | Other |

La columna `JobSatPoints_16` sera eliminada para evitar la multicolinealidad

Asociamos valores 0 (Ninguna intención de cambio) y 1 (Intencion de cambio) para cada grupo dentro de la columna `NewRole`.

Se consideran como validas para el analisis todas las filas que tengan las variables `NewRole` y `JobSatPoints_*` como no nulas. El ranking debe estar completo, con respuestas numericas de 1 al 15 sin repetir.

```python
import pandas as pd

# Se asume que df_encuestas contiene los datos de entrada.
columnas_ranking = sorted(
    [
        columna for columna in df_encuestas.columns
        if columna.startswith("JobSatPoints_")
        and columna.rsplit("_", 1)[-1].isdigit()
    ],
    key=lambda columna: int(columna.rsplit("_", 1)[-1]),
)

# Conservar solo los 15 rangos completos (permutación de 1 a 15) y las cinco
# respuestas válidas de NewRole; JobSatPoints_15_TEXT no forma parte del ranking.
CATEGORIAS_CAMBIO = {
    "I have strongly considered changing my career and/or the industry I work in": 1,
    "I have somewhat considered changing my career and/or the industry I work in": 1,
    "I have transitioned into a new career and/or industry voluntarily": 1,
    "I have neither consider or transitioned into a new career or industry": 0,
    "I have transitioned into a new career and/or industry involuntarily": 0,
}

columnas_requeridas = ["NewRole", *columnas_ranking]
df_modelo = df_encuestas.loc[:, columnas_requeridas].copy()
df_modelo[columnas_ranking] = df_modelo[columnas_ranking].apply(
    pd.to_numeric, errors="coerce"
)
ranking_valido = (
    df_modelo[columnas_ranking].notna().all(axis=1)
    & df_modelo[columnas_ranking].ge(1).all(axis=1)
    & df_modelo[columnas_ranking].le(15).all(axis=1)
    & df_modelo[columnas_ranking].nunique(axis=1).eq(15)
    & df_modelo[columnas_ranking].sum(axis=1).eq(120)
)
newrole_valido = df_modelo["NewRole"].isin(CATEGORIAS_CAMBIO)
df_modelo = df_modelo.loc[ranking_valido & newrole_valido].copy()
df_modelo["y"] = df_modelo["NewRole"].map(CATEGORIAS_CAMBIO).astype(int)

print(f"Observaciones válidas: {len(df_modelo):,}")
```

Luego de extraidas todas las filas validas (aproximadamente el 50% del dataset), procedemos a extraer el ultimo rango para evitar redundancia. Como los 15 rangos siempre suman 120, conociendo 14 se puede deducir el restante. 

Mostramos la distribucion de los dos grupos de la variable dependiente

```python
columnas_predictoras = columnas_ranking[:-1]
X = df_modelo[columnas_predictoras].astype(float)
y = df_modelo["y"]
print("Distribución de y:")
print(y.value_counts().sort_index().rename(index={
    0: "Sin intencion de cambio",
    1: "Con intencion de cambio",
}))
```

La distribucion del subconjunto valido quedo divida proporcionalmente, 48% para el grupo sin intencion de cambio y 52% con intencion de cambio. 

Ahora pasamos al particionamiento de datos previo al entrenamiento de validacion cruzada. 70% para entrenamiento y 30% para test.

Para validar que el muestreo estratificado haya mantenido la misma proporción de clases tanto en el conjunto de entrenamiento (`y_train`) como en el de test (`y_test`), utilizamos un gráfico de barras comparativo mediante `matplotlib`.

```python
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

train_class_counts = pd.Series(y_train).value_counts()
test_class_counts = pd.Series(y_test).value_counts()

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(train_class_counts.index, train_class_counts.values, color="skyblue")
axes[0].set_title("Cantidad de muestras por clase (Entrenamiento)")
axes[0].set_xlabel("Sin intencion de cambio / Con intencion de cambio")
axes[0].set_ylabel("Cantidad de muestras")
axes[0].set_xticks([0,1])

axes[1].bar(test_class_counts.index, test_class_counts.values, color="salmon")
axes[1].set_title("Cantidad de muestras por clase (Test)")
axes[1].set_xlabel("Sin intencion de cambio / Con intencion de cambio")
axes[1].set_ylabel("Cantidad de muestras")
axes[1].set_xticks([0,1])

plt.tight_layout()
plt.show()
```

Estadarizamos los predictores (media 0 y desviación estándar 1) y configuramos el modelo. El algoritmo utiliza automáticamente la penalización L2 (Ridge).

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression 

scaler = StandardScaler()

scaler.fit(X_train)
X_train_stand = scaler.transform(X_train)
X_test_stand = scaler.transform(X_test)

logistic_model = LogisticRegression(max_iter=2000, random_state=42)
```

Ejecutamos el procedimiento de selección de modelo evaluando cada valor de la grilla mediante 5-fold Cross-Validation sobre el subconjunto de entrenamiento estandarizado. En cada paso se mide el promedio de accuracy.

El hiperparámetro C es el inverso de la fuerza de penalización L2 (1/lambda), por lo que valores más pequeños implican una mayor regularización. Esto restringe la magnitud de los pesos y reduce el riesgo overfitting, mejorando la capacidad del modelo para generalizar ante datos no vistos.

Imprimimos el resultado con el hiperparametro optimo y extraemos la instancia del mejor estimador entrenado.

```python
from sklearn.model_selection import GridSearchCV

param_grid = {'C': [0.001, 0.01, 0.1, 1, 10, 100]}

grid_search = GridSearchCV(logistic_model, param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train_stand, y_train)

print("Mejores hiperparámetros (C):", grid_search.best_params_) 
print(f"Mejor exactitud en CV: {grid_search.best_score_:.4f}")

mejor_modelo = grid_search.best_estimator_         
```

La búsqueda determinó que $C = 0.01$ fue la configuración óptima. Sin embargo, la exactitud alcanzada en validación cruzada fue de ~52,7%, lo que indica que, si bien la penalización evita el sobreajuste, los predictores no contienen una relación lineal suficiente para discriminar la intención de cambio por encima del azar.

```python
# El modelo se entrenó con los datos estandarizados en la celda anterior.
from sklearn.metrics import accuracy_score

y_train_pred = mejor_modelo.predict(X_train_stand)
y_test_pred = mejor_modelo.predict(X_test_stand)

print(f"Accuracy CV: {grid_search.best_score_:.3f}")
print(f"Accuracy Train: {accuracy_score(y_train, y_train_pred):.3f}")
print(f"Accuracy Test:  {accuracy_score(y_test, y_test_pred):.3f}")
```

Comprobamos que el modelo no sufre de overfitting; en los tres test la exactitud es similar. De igual forma, estos porcentajes dejan en evidencia la carencia de relacion entre las variables independientes y la dependiente.

Para tener una mejor visualizacion de este problema veamos la matriz de confusion.

```python
import seaborn as sns
from sklearn.metrics import confusion_matrix

matriz = confusion_matrix(y_test, y_test_pred, labels=[0, 1])
plt.figure(figsize=(7, 5))
sns.heatmap(
    matriz,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Predicho: 0", "Predicho: 1"],
    yticklabels=["Real: 0", "Real: 1"],
)
plt.title("Matriz de confusión — Test")
plt.xlabel("Clase predicha")
plt.ylabel("Clase real")
plt.tight_layout()
plt.show()
```

Los errores de tipo 1 y tipo 2 llegan a un número cercano a los aciertos de Sin intención de cambio y Con intención de cambio respectivamente. El modelo realiza predicciones aleatorias. Realizar una exploración de modelos no lineales como ensambles o transformaciones de características podría devolver mejores resultados.

Con los datos expuestos hasta ahora, aceptamos la hipótesis nula ($H_0$) y concluimos que la ausencia de relación lineal entre las variables nos obliga a desechar el modelo de regresión logistica binario para predecir el valor de `NewRole` a partir del conjunto de variables `JobSatPoints_*`.

```python
coeficientes = pd.Series(
    mejor_modelo.coef_.ravel(),
    index=columnas_predictoras,
    name="Coeficiente logístico",
).sort_values(ascending=True)

plt.figure(figsize=(9, 6))
coeficientes.plot(
    kind="barh",
    color=["#c44e52" if valor < 0 else "#4c72b0" for valor in coeficientes],
)
plt.axvline(0, color="black", linewidth=0.8)
plt.title("Coeficientes de la regresión logística binaria")
plt.xlabel("Cambio en log-odds de y=1 por desviación estándar")
plt.ylabel("Aspecto de satisfacción laboral")
plt.tight_layout()
plt.show()
```
