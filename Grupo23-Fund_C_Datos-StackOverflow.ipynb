```python
# Agrupamos HaveWorked: una sola columna con todos los ítems, separados por ';'
# - Celdas NaN se ignoran
# - Cada celda puede traer varios ítems separados por ';' (se parten y se recortan espacios)
HaveWorked_cols = [
    "LanguageHaveWorkedWith",
    "DatabaseHaveWorkedWith",
    "PlatformHaveWorkedWith",
    "WebframeHaveWorkedWith",
    "DevEnvsChoice",
    "DevEnvsHaveWorkedWith",
]


def _tokens_desde_celda(valor):
    """Lista de strings no vacíos; NaN u otro faltante -> lista vacía."""
    if pd.isna(valor):
        return []
    texto = str(valor).strip()
    if not texto:
        return []
    return [parte.strip() for parte in texto.split(";") if parte.strip()]


def _have_worked_agregado(fila):
    ordenados = []
    vistos = set()
    for col in HaveWorked_cols:
        for tok in _tokens_desde_celda(fila[col]):
            if tok not in vistos:
                vistos.add(tok)
                ordenados.append(tok)
    return ";".join(ordenados) if ordenados else pd.NA


df_interes["HaveWorkedWith"] = df_interes[HaveWorked_cols].apply(
    _have_worked_agregado, axis=1
)
df_interes[['HaveWorkedWith']]
```

## Hipótesis — cátedra

#### Hipótesis 1: «Los desarrolladores que respondieron la encuesta en Argentina tienen un salario tres veces mayor que el Salario Mínimo, Vital y Móvil (SMVM)»

Esta hipótesis permite evaluar el subconjunto de encuestados de Argentina y situar los perfiles de IT, tecnología y desarrollo de software frente a un parámetro nacional como el SMVM. El contraste sirve para describir la situación del sector y para comparar un sueldo individual con ese umbral.

Al 1/9/2026, el SMVM es de $383.000. El triple de ese valor es $1.149.000.

Se emplearía un test t de Student si se verificara lo siguiente:

- La variable posee una distribución normal, lo que se evalúa con la prueba de Shapiro-Wilk.
- La variable es cuantitativa continua y está en escala. Esta condición se cumple.
- Independencia: las observaciones de cada desarrollador son independientes.

- H0: μ(salario) ≥ 1.149.000
- H1: μ(salario) < 1.149.000

El contraste es univariado: el salario de esa población frente a un umbral que no está en el CSV y que se obtuvo de datos del Ministerio de Capital Humano de la Nación Argentina (https://www.argentina.gob.ar/trabajo/consejodelsalario).

`Country` permite filtrar las respuestas de Argentina (222 filas). El SMVM es mensual y está expresado en pesos. `CompTotal` (QID22) es la compensación anual, en la moneda de `Currency` (QID21). El salario mensual comparable es `CompTotal / 12` cuando la moneda es el peso argentino (ARS); en caso contrario, se aplica una transformación.
https://www.argentina.gob.ar/trabajo/consejodelsalario
De las 222 filas, 142 declararon `ARS` (`Argentine peso`) y 80, `USD United States dollar`. Hay `CompTotal` en 164 filas (93 en pesos y 71 en dólares) y `ConvertedCompYearly` en 160 (90 y 70). En las 90 filas en pesos que tienen ambos montos, `CompTotal / ConvertedCompYearly` da una mediana de 1.172,26 ARS por USD: es la conversión que Stack Overflow aplicó el 25 de junio de 2025. Para quien declaró dólares, el mensual en pesos es `ConvertedCompYearly × 1.172,26 / 12`. `ConvertedCompYearly` permite auditar esa cuenta.

Población del contraste: `Country = Argentina` y `CompTotal` por encima del piso. Si el monto está en pesos, debe superar 10.000. Si está en dólares, debe superar 10. Son pisos bajos: por debajo de esos valores, la paga anual es demasiado pequeña. Tras ese filtro quedan 160 filas. Se efectuara una sustracciion de los nulos por la mediana , admitido siempre y cuando no modifique la distribución.

```python
# calculamos la mediana
sueldo_mediana_arg = df_arg["Salario_Mensual_USD"].median()
print(f"La mediana de los sueldos de desarrolladores argentinos es: {sueldo_mediana_arg}")
```

```python
from scipy import stats
#ahora si evaluamos normalidad mediante un test de shapiro( n>50)

alfa = 0.05
w, p_valor = stats.shapiro(salario)
print(f"W = {w:.4f}")
print(f"p = {p_valor:.3e}")
print("Se rechaza la normalidad." if p_valor < alfa else "No se rechaza la normalidad.")
```

Al no verificarse la normalidad, se recurre a una prueba no paramétrica: Wilcoxon de rangos con signo para una muestra.

- H0: la mediana del salario mensual de los desarrolladores argentinos es mayor o igual que la cota de 3 × SMVM.
- H1: la mediana es menor que la cota de 3 × SMVM.

```python
from scipy.stats import wilcoxon

salario = df_arg["Salario_Mensual_USD"].dropna()
W, p_valor = wilcoxon(
    salario - SMVM_USD ,
    alternative="less",  # H1: salario por debajo de la cota
    method="auto",
)    
print(f"W = {W:.4f}")
print(f"p = {p_valor:.3e}")
print("Se rechaza la hipotesis, LA mediana de dichos desarrolladores es menor a la cota de SMVM(x3)" if p_valor < alfa else "Efectivamente el salario de los desarrolladores supera la cota de SMVM(x3)")

```

Hemos verificado que no hay evidencia estadística de que la mediana del salario mensual (USD) sea inferior a la cota de 3 × SMVM (aproximadamente 980 USD). Los datos son compatibles con que la mayoría de quienes declararon compensación cobra por encima de ese umbral.
