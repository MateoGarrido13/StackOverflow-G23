## CATEDRA — limpieza de una copia

Este bloque queda aparte de la exploración de arriba. No modifica `df_encuestas`. Arma `df_catedra`: una copia con las columnas de las hipótesis a, b y c, y con los valores que esas hipótesis pueden usar. La celda siguiente conserva el procedimiento completo de las tres hipótesis como guía y no se ejecuta. Debajo se implementa la limpieza de a una, empezando por la (a).

La copia se queda con una fila si es válida para al menos una hipótesis. `valida_a`, `valida_b` y `valida_c` distinguen cada población.

- **a.** `Country = Argentina` y compensación anual mayor a 0. El 0 no es un salario: se anula `CompTotal` y, si lo acompaña, `ConvertedCompYearly`. El mensual en pesos queda en `salario_mensual_ars`, que no viene en el CSV. En pesos es `CompTotal / 12`. En dólares es `ConvertedCompYearly × tasa / 12`. La tasa es la mediana de `CompTotal / ConvertedCompYearly` en las filas de Argentina declaradas en pesos. El umbral de tres SMVM al 1/9/2026 ($1.149.000) se deja anotado y no se usa para tirar filas.
- **b.** `WorkExp` y `AIAcc` con dato. `WorkExp` fuera de 1 a 100 se anula.
- **c.** Las 15 columnas `JobSatPoints_*` son una permutación de los puestos 1 a 15, y `NewRole` es una de las cinco respuestas del cuestionario. No entran `JobSatPoints_15_TEXT` ni `JobSat`.

#### 3. Piso del salario anual

Un solo filtro. Si `CompTotal` está en pesos, tiene que superar 10.000. Si está en dólares, tiene que superar 10. Son cotas bajas: por debajo de eso la paga anual es demasiado chica. Quien no declaró monto, o quedó por debajo, sale de la copia.


````python
PISO_COMPTOTAL_ARS = 10_000
PISO_COMPTOTAL_USD = 10
en_pesos = df_catedra_a["Currency"].str.startswith("ARS", na=False)
en_dolares = df_catedra_a["Currency"].str.startswith("USD", na=False)
supera_piso = (en_pesos & df_catedra_a["CompTotal"].gt(PISO_COMPTOTAL_ARS)) | (
    en_dolares & df_catedra_a["CompTotal"].gt(PISO_COMPTOTAL_USD)
)

print(f"en pesos, con monto y por debajo de {PISO_COMPTOTAL_ARS}: {int((en_pesos & df_catedra_a['CompTotal'].notna() & ~supera_piso).sum())}")
print(f"en dólares, con monto y por debajo de {PISO_COMPTOTAL_USD}: {int((en_dolares & df_catedra_a['CompTotal'].notna() & ~supera_piso).sum())}")
print(f"sin CompTotal: {int(df_catedra_a['CompTotal'].isna().sum())}")

df_catedra_a = df_catedra_a.loc[supera_piso].copy()

print(f"df_catedra_a: {len(df_catedra_a)} filas")
print(df_catedra_a["Currency"].value_counts(dropna=False))

df_catedra_a 
````

````python
# Agrupamos HaveWorked: una sola columna con todos los ítems, separados por ';'
# - Celdas NaN se ignoran
# - Cada celda puede traer varios ítems separados por ';' (se parten y se recortan espacios)
HaveWorked_cols = [
    "LanguageHaveWorkedWith",
    "DatabaseHaveWorkedWith",
    "PlatformHaveWorkedWith",
    "WebframeHaveWorkedWith",
    "DevEnvsChoice",
    "DevEnvsHaveWorkedWith",
]


def _tokens_desde_celda(valor):
    """Lista de strings no vacíos; NaN u otro faltante -> lista vacía."""
    if pd.isna(valor):
        return []
    texto = str(valor).strip()
    if not texto:
        return []
    return [parte.strip() for parte in texto.split(";") if parte.strip()]


def _have_worked_agregado(fila):
    ordenados = []
    vistos = set()
    for col in HaveWorked_cols:
        for tok in _tokens_desde_celda(fila[col]):
            if tok not in vistos:
                vistos.add(tok)
                ordenados.append(tok)
    return ";".join(ordenados) if ordenados else pd.NA


df_interes["HaveWorkedWith"] = df_interes[HaveWorked_cols].apply(
    _have_worked_agregado, axis=1
)
df_interes[['HaveWorkedWith']]
````

````python
# Selección de los 5 JobSatPoints con mejor desempeño general
# En esta escala, un valor menor indica que ese factor es más importante para la satisfacción laboral.
# Por eso se prioriza el promedio más bajo.

PUESTOS_SATISFACCION = [
    "JobSatPoints_1", "JobSatPoints_2", "JobSatPoints_3", "JobSatPoints_4",
    "JobSatPoints_5", "JobSatPoints_6", "JobSatPoints_7", "JobSatPoints_8",
    "JobSatPoints_9", "JobSatPoints_10", "JobSatPoints_11", "JobSatPoints_13",
    "JobSatPoints_14", "JobSatPoints_15", "JobSatPoints_16",
]

job_sat = df_encuestas.loc[:, PUESTOS_SATISFACCION].copy()
for columna in PUESTOS_SATISFACCION:
    job_sat[columna] = pd.to_numeric(job_sat[columna], errors='coerce')

ranking = pd.DataFrame({
    'mean_rank': job_sat.mean(),
    'median_rank': job_sat.median(),
    'n_valid': job_sat.notna().sum(),
    'share_top3': (job_sat <= 3).mean(),
    'share_top5': (job_sat <= 5).mean(),
}).sort_values(['mean_rank', 'median_rank', 'share_top5'], ascending=[True, True, False])

TOP5_JOBSAT = ranking.head(5).index.tolist()
print('Top 5 JobSatPoints por importancia promedio:')
display(ranking.head(5))
print(f'\nColumnas seleccionadas: {TOP5_JOBSAT}')
````
