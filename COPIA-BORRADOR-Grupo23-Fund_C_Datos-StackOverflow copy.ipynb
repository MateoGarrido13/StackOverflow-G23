````python
# Agrupamos HaveWorked: una sola columna con todos los ítems, separados por ';'
# - Celdas NaN se ignoran
# - Cada celda puede traer varios ítems separados por ';' (se parten y se recortan espacios)
HaveWorked_cols = [
    "LanguageHaveWorkedWith",
    "DatabaseHaveWorkedWith",
    "PlatformHaveWorkedWith",
    "WebframeHaveWorkedWith",
    "DevEnvsChoice",
    "DevEnvsHaveWorkedWith",
]


def _tokens_desde_celda(valor):
    """Lista de strings no vacíos; NaN u otro faltante -> lista vacía."""
    if pd.isna(valor):
        return []
    texto = str(valor).strip()
    if not texto:
        return []
    return [parte.strip() for parte in texto.split(";") if parte.strip()]


def _have_worked_agregado(fila):
    ordenados = []
    vistos = set()
    for col in HaveWorked_cols:
        for tok in _tokens_desde_celda(fila[col]):
            if tok not in vistos:
                vistos.add(tok)
                ordenados.append(tok)
    return ";".join(ordenados) if ordenados else pd.NA


df_interes["HaveWorkedWith"] = df_interes[HaveWorked_cols].apply(
    _have_worked_agregado, axis=1
)
df_interes[['HaveWorkedWith']]
````

````python
# Construir cinco columnas con el aspecto que cada persona ubicó en cada puesto del top 5.
# Cada valor es el sufijo del JobSatPoints original, no el puesto del ranking.

PUESTOS_SATISFACCION = [
    "JobSatPoints_1", "JobSatPoints_2", "JobSatPoints_3", "JobSatPoints_4",
    "JobSatPoints_5", "JobSatPoints_6", "JobSatPoints_7", "JobSatPoints_8",
    "JobSatPoints_9", "JobSatPoints_10", "JobSatPoints_11", "JobSatPoints_13",
    "JobSatPoints_14", "JobSatPoints_15", "JobSatPoints_16",
]

job_sat = df_encuestas.loc[:, PUESTOS_SATISFACCION].copy()
for columna in PUESTOS_SATISFACCION:
    job_sat[columna] = pd.to_numeric(job_sat[columna], errors='coerce')

ranking_completo = job_sat.notna().all(axis=1)
ranking_valido = (
    ranking_completo
    & job_sat.ge(1).all(axis=1)
    & job_sat.le(15).all(axis=1)
    & job_sat.nunique(axis=1).eq(len(PUESTOS_SATISFACCION))
)

COLUMNAS_TOP5_JOBSAT = [f"JobSatTop{puesto}" for puesto in range(1, 6)]
ASPECTOS_JOBSAT = [int(columna.rsplit('_', 1)[1]) for columna in PUESTOS_SATISFACCION]
job_sat_top5 = pd.DataFrame(pd.NA, index=job_sat.index, columns=COLUMNAS_TOP5_JOBSAT, dtype='Int64')
indices_ordenados = job_sat.loc[ranking_valido].to_numpy().argsort(axis=1)[:, :5]
job_sat_top5.loc[ranking_valido, :] = [
    [ASPECTOS_JOBSAT[indice] for indice in fila]
    for fila in indices_ordenados
]

print(f'Respuestas con ranking completo y valido: {ranking_valido.sum()} de {len(job_sat)}')
print('Ejemplos de top 5 por persona:')
display(job_sat_top5.loc[ranking_valido].head())
````

- **H0:** los rangos `JobSatPoints_*`, considerados conjuntamente, no se asocian con `NewRole` (sus coeficientes lineales y cuadráticos son conjuntamente cero).
- **H1:** al menos un rango se asocia con `NewRole`.
- **Nivel de significación:** α = 0,05.

````python
# Regresión logística multinomial para evaluar la asociación conjunta.
import warnings

import numpy as np
import pandas as pd
import statsmodels.api as sm
from scipy.stats import chi2

ALFA_LOGIT = 0.05
VARIABLES_LOGIT = PUESTOS_SATISFACCION[:-1]
VARIABLE_OMITIDA = PUESTOS_SATISFACCION[-1]

# Cada observación debe tener un ranking completo: permutación de 1 a 15.
df_modelo = df_analisis.loc[:, ["NewRole", *PUESTOS_SATISFACCION]].copy()
df_modelo[PUESTOS_SATISFACCION] = df_modelo[PUESTOS_SATISFACCION].apply(
    pd.to_numeric, errors="coerce"
)
ranking_valido_modelo = (
    df_modelo[PUESTOS_SATISFACCION].notna().all(axis=1)
    & df_modelo[PUESTOS_SATISFACCION].ge(1).all(axis=1)
    & df_modelo[PUESTOS_SATISFACCION].le(15).all(axis=1)
    & df_modelo[PUESTOS_SATISFACCION].nunique(axis=1).eq(15)
    & df_modelo[PUESTOS_SATISFACCION].sum(axis=1).eq(120)
)
mascara_modelo = ranking_valido_modelo & df_modelo["NewRole"].notna()
df_modelo = df_modelo.loc[mascara_modelo].copy()

print(f"Filas en df_analisis: {len(df_analisis):,}")
print(f"Rankings completos y válidos: {int(ranking_valido_modelo.sum()):,}")
print(f"Casos incluidos (ranking válido + NewRole): {len(df_modelo):,}")
conteos_grupo = df_modelo["NewRole"].value_counts()
display(conteos_grupo.rename_axis("NewRole").to_frame("n"))
print(f"Relación entre el grupo mayor y el menor: {conteos_grupo.max() / conteos_grupo.min():.1f}:1")

ids_modelo = df_encuestas.loc[df_modelo.index, "ResponseId"]
duplicados_id = int(ids_modelo.duplicated().sum())
print(f"ResponseId duplicados: {duplicados_id}")
if duplicados_id:
    raise ValueError("La independencia entre filas no se sostiene: ResponseId duplicados.")
if df_modelo["NewRole"].nunique() != 5:
    raise ValueError("Se esperaban cinco categorías no vacías de NewRole.")

# Los 15 rangos suman 120; se omite uno para evitar colinealidad perfecta.
X_rangos = df_modelo[VARIABLES_LOGIT].astype(float)
media_rangos = X_rangos.mean()
desvio_rangos = X_rangos.std(ddof=0)
if desvio_rangos.eq(0).any():
    raise ValueError("Hay un rango constante que no puede usarse como predictor.")
X_estandarizado = (X_rangos - media_rangos) / desvio_rangos
X_lineal = sm.add_constant(X_estandarizado, has_constant="add")
X_cuadratico = pd.DataFrame(
    {f"{variable}_cuadrado": X_estandarizado[variable] ** 2 for variable in VARIABLES_LOGIT},
    index=df_modelo.index,
)
X_completo = pd.concat([X_lineal, X_cuadratico], axis=1)
rango_diseno = int(np.linalg.matrix_rank(X_completo.to_numpy()))
condicion_diseno = float(np.linalg.cond(X_completo.to_numpy()))
print(f"Diseño cuadrático: {X_completo.shape[1]} columnas; rango {rango_diseno}/{X_completo.shape[1]}; condición {condicion_diseno:.2f}")
if rango_diseno < X_completo.shape[1]:
    raise ValueError("La matriz de diseño no tiene rango completo.")

# NewRole es nominal; se informa el orden y la categoría base usada por MNLogit.
categorias_newrole = sorted(df_modelo["NewRole"].unique())
y = pd.Categorical(df_modelo["NewRole"], categories=categorias_newrole).codes
print("Codificación de NewRole (categoría 0 es la referencia):")
display(pd.DataFrame({"código": range(len(categorias_newrole)), "NewRole": categorias_newrole}))
X_nulo = pd.DataFrame({"const": 1.0}, index=df_modelo.index)

with warnings.catch_warnings(record=True) as avisos_modelo:
    warnings.simplefilter("always")
    ajuste_nulo = sm.MNLogit(y, X_nulo).fit(method="newton", maxiter=100, disp=False)
    ajuste_lineal = sm.MNLogit(y, X_lineal).fit(
        method="newton", maxiter=250, tol=1e-8, disp=False
    )
    ajuste_completo = sm.MNLogit(y, X_completo).fit(
        method="newton", maxiter=300, tol=1e-8, disp=False
    )

convergio_nulo = bool(ajuste_nulo.mle_retvals.get("converged", False))
convergio_lineal = bool(ajuste_lineal.mle_retvals.get("converged", False))
convergio_completo = bool(ajuste_completo.mle_retvals.get("converged", False))
parametros_finitos = bool(np.isfinite(ajuste_completo.params.to_numpy()).all())
se_completos = ajuste_completo.bse.to_numpy()
errores_estandar_finitos = bool(np.isfinite(se_completos).all())
error_estandar_maximo = float(np.max(se_completos))
probabilidad_maxima = float(np.asarray(ajuste_completo.predict(X_completo)).max())
print(f"Convergencia nulo/lineal/cuadrático: {convergio_nulo}/{convergio_lineal}/{convergio_completo}")
print(f"Coeficientes finitos: {parametros_finitos}; error estándar máximo: {error_estandar_maximo:.3f}")
print(f"Probabilidad predicha máxima: {probabilidad_maxima:.4f}")
print(f"Avisos del ajuste: {[str(aviso.message) for aviso in avisos_modelo]}")

if convergio_lineal and convergio_completo:
    lr_curvatura = 2 * (ajuste_completo.llf - ajuste_lineal.llf)
    gl_curvatura = int(ajuste_completo.params.size - ajuste_lineal.params.size)
    p_curvatura = float(chi2.sf(lr_curvatura, gl_curvatura))
    print(f"Prueba de términos cuadráticos vs. lineales: LR={lr_curvatura:.2f}, gl={gl_curvatura}, p={p_curvatura:.3g}")
    if p_curvatura < ALFA_LOGIT:
        print("Los términos cuadráticos mejoran el ajuste frente al modelo solo lineal.")
else:
    p_curvatura = None
    print("No se interpreta la prueba lineal vs. cuadrática: alguno de los modelos no convergió.")

if (
    convergio_nulo and convergio_completo and parametros_finitos
    and errores_estandar_finitos
):
    estadistico_lr = 2 * (ajuste_completo.llf - ajuste_nulo.llf)
    grados_libertad = int(ajuste_completo.params.size - ajuste_nulo.params.size)
    if estadistico_lr < -1e-6:
        raise ValueError("El modelo completo tuvo menor verosimilitud que el nulo; revisar la especificación.")
    estadistico_lr = max(0.0, estadistico_lr)
    p_valor_global = float(chi2.sf(estadistico_lr, grados_libertad))
    pseudo_r2 = float(1 - ajuste_completo.llf / ajuste_nulo.llf)
    resultado_global = pd.DataFrame({
        "LR chi2": [estadistico_lr],
        "gl": [grados_libertad],
        "p-valor": [p_valor_global],
        "Pseudo-R2 McFadden": [pseudo_r2],
    }, index=["14 rangos (términos lineales + cuadráticos)"])
    display(resultado_global)
    if p_valor_global < ALFA_LOGIT:
        print("Decisión (α=0,05): rechazar H0; hay evidencia de asociación entre el perfil de rangos y NewRole.")
    else:
        print("Decisión (α=0,05): no rechazar H0; no hay evidencia suficiente de asociación.")
else:
    resultado_global = None
    print("No se reporta el contraste global: hay falta de convergencia o estimaciones no finitas.")

print("La regresión logística no requiere normalidad ni homocedasticidad de los predictores.")
print("IIA es un supuesto del logit multinomial y no se verifica automáticamente en esta celda.")
print(f"Nota: se excluyen del modelo {VARIABLE_OMITIDA} y su término cuadrático porque el rango y los cuadrados de los 15 puestos tienen sumas constantes (120 y 1240); df_analisis conserva los 15 rangos.")
````

## BORRADOR 9/11 — Comparativa de la hipótesis 3

Comparamos tres formas de anticipar si un desarrollador considera un cambio de rol o de carrera. La primera es la regresión logística binaria ya ajustada en la hipótesis 3 de la cátedra. La segunda es predecir siempre la clase mayoritaria. La tercera es un árbol de decisión de profundidad 3: cada `JobSatPoints_*` se parte con cota 5, de modo que un puesto mayor que 5 cuenta como aspecto prioritario y un puesto de 5 o menos, como no prioritario.

El punto de partida es la matriz de confusión de la logística en el test. Al lado va la matriz de quien predice siempre la mayoría. Esas dos lecturas sirven para evaluar el árbol.

El criterio de la comparación es el error de tipo 2: predecir que la persona no quiere rotar cuando en realidad sí. Para una empresa, no detectar a tiempo a un desarrollador que quiere cambiar de rol o de carrera es el riesgo que se busca achicar. El error de tipo 1 es predecir un cambio cuando la persona no lo consideró ni transitó.

La logística se rearma igual que en la hipótesis ya desarrollada. `y = 0` queda solo para quien no consideró ni transitó un cambio. Considerar el cambio, o haberlo hecho de forma voluntaria o involuntaria, es `y = 1`. Entran las filas con el ranking completo de 1 a 15. `JobSatPoints_16` se omite en la logística porque los quince puestos suman 120. El split es 70/30, estratificado, con `random_state=42`. El `C` es 10, el valor que eligió la grilla. El árbol usa el mismo reparto de filas y las quince columnas ya cortadas en la cota 5.

### 2. Matriz de confusión de la regresión logística

Se reconstruye el subconjunto de la hipótesis 3 y se ajusta la logística sobre el entrenamiento estandarizado. La matriz es la del test: filas reales y columnas predichas, con 0 = sin intención de cambio y 1 = con intención de cambio.

````python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier

columnas_ind_h3 = sorted(
    [
        columna for columna in df_encuestas.columns
        if columna.startswith("JobSatPoints_")
        and columna.rsplit("_", 1)[-1].isdigit()
    ],
    key=lambda columna: int(columna.rsplit("_", 1)[-1]),
)

# 0 es únicamente quien no consideró ni transitó. Considerar el cambio, o haberlo
# hecho de forma voluntaria o involuntaria, cuenta como intención de cambio (1).
valores_dep_h3 = {
    "I have strongly considered changing my career and/or the industry I work in": 1,
    "I have somewhat considered changing my career and/or the industry I work in": 1,
    "I have transitioned into a new career and/or industry voluntarily": 1,
    "I have transitioned into a new career and/or industry involuntarily": 1,
    "I have neither consider or transitioned into a new career or industry": 0,
}

df_job_sat = df_encuestas.loc[:, ["NewRole", *columnas_ind_h3]].copy()
df_job_sat[columnas_ind_h3] = df_job_sat[columnas_ind_h3].apply(pd.to_numeric, errors="coerce")
idx_ranking_valido_h3 = (
    df_job_sat[columnas_ind_h3].notna().all(axis=1)
    & df_job_sat[columnas_ind_h3].ge(1).all(axis=1)
    & df_job_sat[columnas_ind_h3].le(15).all(axis=1)
    & df_job_sat[columnas_ind_h3].nunique(axis=1).eq(15)
    & df_job_sat[columnas_ind_h3].sum(axis=1).eq(120)
)
df_job_sat = df_job_sat.loc[
    idx_ranking_valido_h3 & df_job_sat["NewRole"].isin(valores_dep_h3)
].copy()
df_job_sat["var_dep_h3"] = df_job_sat["NewRole"].map(valores_dep_h3).astype(int)

# Los 15 puestos suman 120: la logística omite JobSatPoints_16.
nombres_vars_ind_logistica_h3 = columnas_ind_h3[:-1]
vars_ind_logistica_h3 = df_job_sat[nombres_vars_ind_logistica_h3].astype(float)
var_dep_h3 = df_job_sat["var_dep_h3"]

idx_train_h3, idx_test_h3 = train_test_split(
    df_job_sat.index, test_size=0.30, random_state=42, stratify=var_dep_h3
)
vars_ind_logistica_h3_train = vars_ind_logistica_h3.loc[idx_train_h3]
vars_ind_logistica_h3_test = vars_ind_logistica_h3.loc[idx_test_h3]
var_dep_h3_train = var_dep_h3.loc[idx_train_h3]
var_dep_h3_test = var_dep_h3.loc[idx_test_h3]

scaler_h3 = StandardScaler()
vars_ind_logistica_h3_train_estandar = scaler_h3.fit_transform(vars_ind_logistica_h3_train)
vars_ind_logistica_h3_test_estandar = scaler_h3.transform(vars_ind_logistica_h3_test)

logistica_h3 = LogisticRegression(C=10, max_iter=2000, random_state=42)
logistica_h3.fit(vars_ind_logistica_h3_train_estandar, var_dep_h3_train)
var_dep_h3_pred_logistica = logistica_h3.predict(vars_ind_logistica_h3_test_estandar)


def mostrar_matriz_h3(var_dep_real, var_dep_pred, titulo):
    matriz = confusion_matrix(var_dep_real, var_dep_pred, labels=[0, 1])
    tn, fp, fn, tp = (int(valor) for valor in matriz.ravel())
    n = tn + fp + fn + tp
    n0 = tn + fp
    n1 = fn + tp
    print(f"Acierto: {(tn + tp) / n:.3f}")
    print(f"Test: {n:,} filas ({n0:,} sin intención, {n1:,} con intención)")
    print(
        f"Tipo 1 (predice cambio y no lo hubo): {fp:,} "
        f"({fp / n:.1%} del test; {fp / n0:.1%} de la clase 0)"
    )
    print(
        f"Tipo 2 (predice que no cambia y sí lo consideró): {fn:,} "
        f"({fn / n:.1%} del test; {fn / n1:.1%} de la clase 1)"
    )
    plt.figure(figsize=(7, 5))
    sns.heatmap(
        matriz,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Predicho: 0", "Predicho: 1"],
        yticklabels=["Real: 0", "Real: 1"],
    )
    plt.title(titulo)
    plt.xlabel("Clase predicha")
    plt.ylabel("Clase real")
    plt.tight_layout()
    plt.show()
    return matriz


print(f"Observaciones válidas: {len(df_job_sat):,}")
print(
    var_dep_h3.value_counts()
    .sort_index()
    .rename(index={0: "Sin intencion de cambio", 1: "Con intencion de cambio"})
)
mostrar_matriz_h3(
    var_dep_h3_test, var_dep_h3_pred_logistica, "Regresión logística — Test"
)
````

En el test (7.419 filas: 3.438 sin intención y 3.981 con intención) la logística queda así:

|  | Predicho: sin intención | Predicho: con intención |
|---|---:|---:|
| Real: sin intención | 583 | 2.855 |
| Real: con intención | 499 | 3.482 |

El error de tipo 1 son 2.855 casos: el 38,5% del test y el 83,0% de quienes están en la clase 0. El error de tipo 2 son 499 casos: el 6,7% del test y el 12,5% de quienes quieren un cambio. El modelo predice "con intención" en 6.337 filas, el 85,4% del test. La precisión de esa clase es 54,9% (3.482 / 6.337) y el acierto es 54,8%.

Esa matriz es el punto de partida. La mayoría del test, predicha siempre, acierta el 53,7%. La logística queda alrededor de un punto por encima y, al marcar tan seguido la clase 1, deja pasar 499 desarrolladores que sí consideran rotar.

### 3. Predecir siempre por la mayoría

La clase mayoritaria del entrenamiento es "con intención de cambio" (`y = 1`). Un clasificador trivial que siempre predice esa clase se ajusta solo con `var_dep_h3_train` y se evalúa en el mismo test que la logística. Su acierto es la cota contra la cual se lee cualquier modelo: si no la supera, el ranking de satisfacción no aporta una clasificación distinta de decir siempre que la persona quiere rotar.

````python
mayoria_h3 = DummyClassifier(strategy="most_frequent")
mayoria_h3.fit(vars_ind_logistica_h3_train_estandar, var_dep_h3_train)
var_dep_h3_pred_mayoria = mayoria_h3.predict(vars_ind_logistica_h3_test_estandar)

print(f"Clase mayoritaria del entrenamiento: {int(var_dep_h3_train.value_counts().idxmax())}")
mostrar_matriz_h3(var_dep_h3_test, var_dep_h3_pred_mayoria, "Mayoría — Test")
````

La matriz de la mayoría, en las mismas 7.419 filas, predice cambio en todos los casos:

|  | Predicho: sin intención | Predicho: con intención |
|---|---:|---:|
| Real: sin intención | 0 | 3.438 |
| Real: con intención | 0 | 3.981 |

El acierto es 53,7% (3.981 / 7.419). Dado que siempre predecimos que quiere rotar , no habran falsos negativos, sin embargo puede ser pelgroso un cambio de rol en una persona que estaba conforme en donde estaba (error tipo 1 ), las mismas son las 3.438 personas de la clase 0, el 46,3% del test y el 100% de quienes no consideran un cambio.

### 4. Alternativa: árbol de decisión de profundidad 3 con cota 5

Cada `JobSatPoints_*` pasa a 1 si el puesto es mayor que 5 y a 0 si es 5 o menos. Entran las quince columnas del ranking. El árbol tiene profundidad fija 3, `random_state=42`, y se entrena con las mismas filas de entrenamiento que la logística. El criterio sigue siendo el error de tipo 2: cuántos desarrolladores que quieren rotar quedan sin detectar.

````python
vars_ind_arbol_h3 = (df_job_sat[nombres_vars_ind_h3] > 5).astype(int)
vars_ind_arbol_h3_train = vars_ind_arbol_h3.loc[idx_train_h3]
vars_ind_arbol_h3_test = vars_ind_arbol_h3.loc[idx_test_h3]

arbol_h3 = DecisionTreeClassifier(max_depth=3, random_state=42)
arbol_h3.fit(vars_ind_arbol_h3_train, var_dep_h3_train)
var_dep_h3_pred_arbol = arbol_h3.predict(vars_ind_arbol_h3_test)
probas_hoja_h3 = arbol_h3.predict_proba(vars_ind_arbol_h3_test)[:, 1]

print(f"Hojas: {arbol_h3.get_n_leaves()}")
print(f"Profundidad efectiva: {arbol_h3.get_depth()}")
print(f"Probabilidad mínima de una hoja en el test: {probas_hoja_h3.min():.3f}")
mostrar_matriz_h3(
    var_dep_h3_test,
    var_dep_h3_pred_arbol,
    "Árbol de profundidad 3, cota 5 — Test",
)
````

Con la cota 5 y profundidad 3, la probabilidad más baja de una hoja en el test es 0,503. Todas las hojas quedan del lado de la clase 1, así que el árbol predice "con intención" en las 7.419 filas. Su matriz coincide con la de la mayoría:

|  | Predicho: sin intención | Predicho: con intención |
|---|---:|---:|
| Real: sin intención | 0 | 3.438 |
| Real: con intención | 0 | 3.981 |

El error de tipo 2 es 0, igual que al predecir siempre la mayoría. El error de tipo 1 son las 3.438 personas que no consideran un cambio. El acierto queda en 53,7%.

Bajo la prioridad de no dejar pasar a quien quiere rotar, el árbol y la mayoría empatan en 0 no detectados. La logística deja pasar 499 (el 12,5% de quienes sí consideran un cambio) y, a cambio, reconoce a 583 personas de la clase 0. El corte en 5 con un árbol de tres niveles repite la regla trivial: marca a todo el test.

Ese empate con “predecir siempre la mayoría” sale de las hojas. En las ocho, el valor de “sí” (clase 1, con intención) es mayor que el de “no”: la proporción de “sí” va de 0,503 a 0,604. Cada hoja predice la clase con más casos, así que todas responden “con intención” y el árbol marca las 7.419 filas del test, igual que la regla de la mayoría.

El árbol, además, elige el orden de las preguntas. En cada nodo se queda con la que más divide el grupo, la que más separa “sí” y “no”. Ese orden se puede leer como un preprocesamiento: de las quince columnas ya cortadas en la cota 5, el modelo filtra por la pregunta de mayor división y después encadena las siguientes del mismo modo. Con profundidad 3 la primera es `JobSatPoints_4`. El filtro ordena las preguntas; en este ajuste, cada hoja sigue teniendo un “sí” mayor.
